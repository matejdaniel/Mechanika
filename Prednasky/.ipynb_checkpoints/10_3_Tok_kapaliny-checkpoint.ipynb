## Popis proudění (Kinematika tekutin)

V mechanice pevných těles obvykle popisujeme pohyb tak, že sledujeme polohu celého tělesa v čase. 
V mechanice tekutin by tento přístup vyžadoval sledovat dráhu každé jednotlivé částice tekutiny zvlášť, 
což je vzhledem k jejich obrovskému množství teoreticky i prakticky nesmírně obtížné.
Z tohoto důvodu mechanika tekutin definuje dva základní způsoby, jak pohyb kapalin a plynů 
zkoumat a popisovat:

### Lagrangeův přístup (sledování částice)

Lagrangeův přístup je analogický k mechanice pevných těles – vybíráme si jednu konkrétní částici
(pohybující se s proudem) a sledujeme její dráhu v prostoru a čase. 
Rychlostní vektor této sledované částice vyjádříme jako $\vec{v} = u\hat{i} + v\hat{j} + w\hat{k}$.
Pokud známe polohu částice $x_0 = (x_0, y_0, z_0)$ v čase $t = t_0$, 
můžeme matematicky určit její rychlost a následně i zrychlení jako časovou derivaci rychlosti $\vec{a} = \frac{\partial \vec{v}}{\partial t}$. Na takto popsanou částici lze následně přímo aplikovat 2. Newtonův pohybový zákon ($\vec{F} = m\vec{a}$) pro stanovení sil.

### Eulerův přístup (sledování pevného bodu v prostoru)

V laboratorní a technické praxi je mnohem snazší instalovat senzory na jedno místo (např. do potrubí nebo aerodynamického tunelu) a sledovat, jak kolem nich tekutina protéká. K tomu slouží Eulerův přístup. Místo sledování jednotlivých částic definujeme pevný kontrolní objem (nebo bod) v prostoru $(x, y, z)$ a analyzujeme stav tekutiny, která jím v čase $t$ právě prochází. Rychlost je tak popsána jako vektorové pole, které je funkcí polohy a času: $\vec{V}(\vec{x}, t)$.

```{note}
Abychom mohli aplikovat Newtonovy zákony na kontrolní objem protékaný různými částicemi, využívá se speciální matematický aparát zvaný **Reynoldsova transportní věta**, která tvoří přemostění mezi Lagrangeovým (systémovým) a Eulerovým (kontrolním) pojetím.
```

### Křivky k vizualizaci proudění
Pro pochopení a popis toku definujeme v tekutinách tři typy čar:

* **Proudnice (Streamlines):** Křivky, které jsou v každém svém bodě tečné k vektoru rychlosti $\vec{V}$ z Eulerova popisu. Z definice neexistuje žádná rychlostní složka kolmá na proudnici (tekutina proudnici nepřetéká).
* **Stopové čáry (Streaklines):** Okamžitá křivka spojující všechny částice, které v minulosti prošly jedním určitým bodem $x_0$. Lze si ji představit jako stopu barviva, které kontinuálně vstřikujeme do jednoho bodu v proudu.
* **Trajektorie (Pathlines):** Skutečná fyzická dráha, kterou v čase vykreslí jedna sledovaná částice (čistě Lagrangeův pohled).

```{important}
V případě **stacionárního (ustáleného) proudění**, kde se rychlostní pole nemění v čase, jsou proudnice, stopové čáry i trajektorie totožné a dokonale splývají v jednu křivku.
```

## Základní zákony zachování 

Řešení problémů mechaniky tekutin spočívá v aplikaci zákonů zachování na kontrolní objem (CV). Ten funguje jako "fluidní černá skříňka", do které vstupují a ze které vystupují hmota, hybnost a energie. Kontrolní objem může být teoreticky pohyblivý, ale ve většině inženýrských aplikací (i v tomto textu) jej považujeme za pevně fixovaný v prostoru.

V hydrodynamice vždy platí tři základní zákony: Zákon zachování hmotnosti, Zákon zachování hybnosti a Zákon zachování energie.

### 2.1 Zákon zachování hmotnosti (Rovnice kontinuity)

Základním principem je, že celková hmotnost vstupující do kontrolního objemu se musí rovnat hmotnosti z něj vystupující, plus hmotnosti, která se případně uvnitř objemu naakumuluje (nahromadí).

Zapsáno pomocí hmotnostních toků $\dot{m}$:

$$ \dot{m}_{in} - \dot{m}_{out} = \dot{m}_{acc} $$

```{admonition} Fyzikální interpretace na dvou případech
:class: tip

**Případ I: Vana s odtokem**
Sledujeme vanu napouštěnou shora a vypouštěnou spodním odtokem. Přitékající hmota je dána součinem hustoty, rychlosti a plochy přítoku: $\dot{m}_{in} = \rho V_{in} A_{in}$. Pokud do vany přitéká voda rychleji, než z ní odtéká ($\dot{m}_{in} > \dot{m}_{out}$), voda se ve vaně hromadí a hladina stoupá. Pokud je odtok větší, vana se vyprázdní. Rovnováha nastane pouze tehdy, když se hmotnostní toky vyrovnají.

**Případ II: Zcela zaplněné potrubí**

Pokud máme kontrolní objem definovaný jako zaplněný úsek potrubí s nestlačitelnou tekutinou (vodou), její hustota je konstantní. Do objemu, který je již plný, nelze stlačit další hmotu, proto je akumulace nulová ($\dot{m}_{acc} = 0$) a tok dovnitř se musí rovnat toku ven: $\dot{m}_{in} = \dot{m}_{out}$. Pokud má potrubí konstantní průřez, musí platit i to, že rychlost na vstupu je stejná jako na výstupu ($V_{in} = V_{out}$).
```

### Diferenciální kontrolní objem (Odvození 3D rovnice kontinuity)

Uvažujme infinitesimálně malý kontrolní objem o hranách $dx, dy, dz$ ponořený v proudící tekutině.
Na levé stěně vstupuje hmota ve směru osy $x$ rychlostí $u$:

$$ \dot{m}_{in,x} = \rho \, u \, dy \, dz $$

Na protější stěně, vzdálené o $dx$, se rychlost může mírně změnit. Tuto změnu aproximujeme pomocí prvního členu Taylorova rozvoje:

$$ \dot{m}_{out,x} = \rho \left( u + \frac{\partial u}{\partial x} dx \right) dy \, dz $$

Podobně odvodíme vstupy a výstupy pro osu $y$ (rychlost $v$) a osu $z$ (rychlost $w$). Dosadíme do zákona zachování hmotnosti ($\dot{m}_{in} - \dot{m}_{out} = 0$):

$$ (\rho u \, dy \, dz + \rho v \, dx \, dz) - \left[ \rho \left( u + \frac{\partial u}{\partial x} dx \right) dy \, dz + \rho \left( v + \frac{\partial v}{\partial y} dy \right) dx \, dz \right] = 0 $$

Vstupní toky se vykrátí s prvními členy v závorce a dostaneme:

$$ -\rho \frac{\partial u}{\partial x} dx \, dy \, dz - \rho \frac{\partial v}{\partial y} dx \, dy \, dz = 0 $$

Tuto rovnici vydělíme hustotou a elementárním objemem ($dx dy dz$) a získáme diferenciální tvar rovnice kontinuity pro dvoudimenzionální proudění:

$$ \frac{\partial u}{\partial x} + \frac{\partial v}{\partial y} = 0 $$

Při rozšíření do tří dimenzí a využití operátoru divergence (nabla $\nabla$) dostaneme elegantní vektorový zápis rovnice kontinuity nestlačitelné tekutiny:

$$ \nabla \cdot \vec{v} = 0 $$.

### Zákon zachování hybnosti (Eulerova rovnice)

Odvození pohybové rovnice pro tekutiny vychází přímo z 2. Newtonova zákona, který říká, že časová změna hybnosti soustavy je rovna součtu působících sil ($\Sigma F_i = \frac{d}{dt}(p)$).

Jak Richard Feynman s oblibou připomínal ve svých slavných *Přednáškách z fyziky*, když aplikujeme Newtonův pohybový zákon pro výpočet síly na tekutinu, musíme být velmi opatrní v tom, co přesně myslíme "zrychlením" tělesa. Naším cílem je popsat zrychlení jedné **konkrétní částice** tekutiny, kterou pomyslně sledujeme na její dráze. My ale obvykle neznáme vzorce pro dráhy jednotlivých částic. Známe "jen" rychlostní pole $\vec{v}(x,y,z,t)$, které nám říká, jak rychle a kam proudí tekutina v **konkrétním pevném bodě** prostoru a v daném čase.

Když se tedy podíváme na částici v proudu, její rychlost se může změnit ze dvou logicky zcela odlišných důvodů:

1. **Lokální změna (v čase):** I kdyby částice stála na jednom místě (což v proudu nelze), rychlost toku v onom bodě se může měnit (například proto, že někdo právě otevírá nebo zavírá ventil na potrubí). Tuto změnu vystihuje parciální derivace rychlosti podle času $\frac{\partial \vec{v}}{\partial t}$
.
2. **Konvektivní / advektivní změna (v prostoru):** I v dokonale ustáleném proudu (kde se v daném místě s časem vůbec nic nemění) se rychlost částice mění. Proč? Protože tekutina proudí třeba do zúžené části trubice. Jak se částice přesune na novou pozici (kde je podle rychlostního pole rychlost vyšší), musí nutně zrychlit.

Představme si jednu konkrétní částici. Její poloha v čase $t$ je dána souřadnicemi $x(t), y(t), z(t)$.
Okamžitá rychlost této částice má složky $v_x, v_y, v_z$. Přímo z definice rychlosti tedy víme, že:

$$\frac{dx}{dt} = v_x, \quad \frac{dy}{dt} = v_y, \quad \frac{dz}{dt} = v_z$$

Zajímejme se teď jen o to, jak se v čase mění $x$-ová složka rychlosti této naší jedoucí částice. Hledáme tedy její úplnou derivaci podle času: $\frac{dv_x}{dt}$.
Protože rychlost v poli závisí na poloze a čase, tedy $v_x = v_x(x,y,z,t)$, musíme použít pravidlo pro derivování složené funkce. Změna rychlosti se poskládá ze změn podél všech proměnných:

$$\frac{dv_x}{dt} = \frac{\partial v_x}{\partial t} + \frac{\partial v_x}{\partial x}\frac{dx}{dt} + \frac{\partial v_x}{\partial y}\frac{dy}{dt} + \frac{\partial v_x}{\partial z}\frac{dz}{dt}$$

Derivace polohy částice (např. $\frac{dx}{dt}$) nejsou nic jiného než přímo složky rychlosti částice v daném směru ($v_x$). Po dosazení dostaneme:

$$\frac{dv_x}{dt} = \frac{\partial v_x}{\partial t} + v_x \frac{\partial v_x}{\partial x} + v_y \frac{\partial v_x}{\partial y} + v_z \frac{\partial v_x}{\partial z}$$

První člen představuje ono lokální zrychlení a zbylé tři členy tvoří **advektivní (konvektivní) zrychlení** ve směru osy $x$.

### Vektorový zápis

Abychom to stejné nemuseli vypisovat třikrát pro osy $x, y$ a $z$, zavedli matematici elegantní vektorový zápis. Využijeme operátor gradientu $\nabla = \left( \frac{\partial}{\partial x}, \frac{\partial}{\partial y}, \frac{\partial}{\partial z} \right)$ a vytvoříme s ním nový skalární diferenciální operátor pomocí skalárního součinu s vektorem rychlosti:

$$(\vec{v} \cdot \nabla) = v_x \frac{\partial}{\partial x} + v_y \frac{\partial}{\partial y} + v_z \frac{\partial}{\partial z}$$

Tento operátor v podstatě říká: *"Vezmi prostorovou změnu ve všech směrech a vynásob ji tím, jak rychle se v těchto směrech pohybujeme."* Pokud tento operátor aplikujeme na samotný vektor rychlosti $\vec{v}$, získáme přesně ty tři prostorové složky z našeho odvození pro celý vektor.

Celkové zrychlení částice tekutiny $\vec{a}$ (v hydrodynamice se pro zdůraznění pohybu s částicí značí jako tzv. materiálová derivace $\frac{D\vec{v}}{Dt}$) je tedy součtem:

$$\vec{a} = \frac{D\vec{v}}{Dt} = \underbrace{\frac{\partial \vec{v}}{\partial t}}_{\text{Lokální zrychl.}} + \underbrace{(\vec{v} \cdot \nabla)\vec{v}}_{\text{Advektivní zrychl.}}$$

Člen $(\vec{v} \cdot \nabla)\vec{v}$ je jádrem veškeré složitosti mechaniky tekutin. Reprezentuje silnou nelinearitu – rychlost proudění je ovlivňována (advektována) sama sebou. Je to právě tento člen, kvůli kterému jsou Navier-Stokesovy rovnice tak obtížně řešitelné a díky kterému vznikají turbulence.

**Síly působící na tekutinu:**
Pro odvození předpokládáme ideální neviskózní tekutinu (zanedbáváme vnitřní tření). V takovém případě na element působí pouze dvě síly:
* **Gravitace:** Tíha elementárního objemu směřující dolů (proti ose $z$). $F_g = -(\rho g \, dx \, dz) \hat{k}$.
* **Gradient tlaku:** Pokud tlak není všude stejný, působí na element tekutiny výsledná tlaková síla směrem z místa vyššího tlaku do nižšího. Obdobně jako u rovnice kontinuity odvodíme rozdíl tlaků na protilehlých stěnách. Tlaková síla je plošný obsah krát gradient tlaku: $F_p = -\nabla p \, dx \, dz$.

**Výsledná rovnice:**
Dáme-li dohromady zrychlení, hustotu a působící síly a vydělíme objemem elementu, získáme **Eulerovou pohybovou rovnici** v její vektorové formě:

$$ \rho \left\{ \frac{\partial \vec{v}}{\partial t} + (\vec{v} \cdot \nabla)\vec{v} \right\} = -\rho \vec{g} - \nabla p $$.

### Navier-Stokesovy rovnice

Eulerova rovnice je zjednodušenou verzí Navier-Stokesových rovnic, ve které zcela ignorujeme viskozitu (vnitřní tření tekutiny). Navier-Stokesovy rovnice **reálnou tekutinu s viskozitou** ($\mu > 0$).

* Rovnice počítají s vnitřním třením.
* Dokážou popsat vznik turbulencí, úbytek tlaku v potrubí vlivem tření a vznik aerodynamického odporu.

 Levá strana (reprezentující zrychlení částice) je u obou rovnic naprosto identická. Rozdíl je na pravé straně (síly):

**Eulerova rovnice:**


$$\rho \frac{D\vec{v}}{Dt} = -\nabla p + \rho \vec{g}$$

**Navier-Stokesova rovnice:**

$$\rho \frac{D\vec{v}}{Dt} = -\nabla p + \mu \nabla^2 \vec{v} + \rho \vec{g}$$

### Bernoulliho rovnice - odvození z Navier-Stokesových rovnic

Bernoulliho rovnice je integrálním výsledkem Eulerovy rovnice řešené podél jedné proudnice a vyjadřuje zákon zachování mechanické energie tekutiny.
```{warning} Podmínky platnosti
Abychom Bernoulliho rovnici mohli v praxi použít, je nutné striktně splnit následujících 5 předpokladů:
1. Aplikuje se pouze na body ležící na **téže proudnici**.
2. Tekutina je **nestlačitelná** (konstantní hustota $\rho$).
3. Proudění je **stacionární** (nemění se v čase a bez turbulencí, $\frac{dv}{dt} = 0$).
4. Tekutina je **ideální (neviskózní)** – nedochází ke ztrátám mechanické energie třením v hraničních vrstvách.
5. Do systému není přidávána ani z něj odebírána vnější mechanická práce (čerpadla, turbíny) a nedodává se teplo.
```

**Odvození podél proudnice:**
Uvažujme válcový element tekutiny na proudnici se směrnicí $s$ a délkou $ds$.
Složka tíhy, která směřuje proti směru toku tekutiny, je určena úhlem sklonu proudnice $\beta$ jako $-\rho g \sin(\beta) dV$. Vzhledem k tomu, že ze zavedené geometrie platí $\sin(\beta) = \frac{dz}{ds}$, můžeme elementární tíhovou sílu zapsat jako $-\rho g \frac{\partial z}{\partial s} \, dn \, ds \, dx$.
Výsledná tlaková síla působící na konce elementu je určena derivací tlaku podél proudnice $-\frac{\partial p}{\partial s} \, dn \, ds \, dx$.

Podle Newtonova zákona musí součet těchto sil odpovídat zrychlení elementu $\rho \vec{v} \frac{\partial \vec{v}}{\partial s}$. Po vydělení elementárním objemem získáváme 1D Eulerovu rovnici podél proudnice:

$$ \rho \vec{v} \frac{\partial \vec{v}}{\partial s} + \frac{\partial p}{\partial s} + \rho g \frac{\partial z}{\partial s} = 0 $$

Převedením parciálních derivací na úplné diferenciály ($dV, dp, dz$) integrací přes úsek $ds$ získáme tvar:

$$ \rho v \, dv + dp + \rho g \, dz = 0 $$

Vydělením konstantní hustotou $\rho$ a následným integrováním od bodu 1 do bodu 2 na proudnici dostaneme finální Bernoulliho rovnici:

$$ \frac{p}{\rho} + \frac{1}{2} v^2 + g z = konst. $$

Tato rovnice spojuje tlakovou ($p/\rho$), kinetickou ($\frac{1}{2}V^2$) a potenciální ($gz$) energii tekutiny do jednoho zachovávajícího se celku. Využívá se ve formátu srovnávání dvou stavů podél proudu:

$$ \frac{p_1}{\rho} + \frac{1}{2} V_1^2 + g z_1 = \frac{p_2}{\rho} + \frac{1}{2} V_2^2 + g z_2 $$.

Pro lepší pochopení toho, jak změna průřezu nebo výšky potrubí ovlivňuje rychlost a tlak kapaliny podle rovnice kontinuity a Bernoulliho rovnice, můžete využít následující interaktivní simulaci.


### Odvození Bernoulliho rovnice ze zákona zachování energie

Bernoulliho rovnice je v podstatě energetická bilance pro ideální tekutinu. Předpokládáme ustálené proudění nestlačitelné tekutiny bez vnitřního tření (neviskózní proudění) podél proudnice. Uvažujme úsek tekutiny v potrubí (proudovou trubici). Sledujeme objem tekutiny, který za malý časový interval $\Delta t$ projde dvěma průřezy:

* **V bodě 1:** Tlak $p_1$, rychlost $v_1$, výška $z_1$ a plocha průřezu $A_1$.
* **V bodě 2:** Tlak $p_2$, rychlost $v_2$, výška $z_2$ a plocha průřezu $A_2$.

Za čas $\Delta t$ se v bodě 1 posune tekutina o dráhu $\Delta s_1 = v_1 \Delta t$ a v bodě 2 o dráhu $\Delta s_2 = v_2 \Delta t$. Objem přenesené tekutiny je díky nestlačitelnosti stejný: $\Delta V = A_1 \Delta s_1 = A_2 \Delta s_2$. Na tekutinu působí okolní tlakové síly.

* **Práce vykonaná na vstupu (bod 1):** Okolí tlačí tekutinu dovnitř silou $F_1 = p_1 A_1$. Práce je kladná:

$$W_1 = F_1 \Delta s_1 = p_1 A_1 \Delta s_1 = p_1 \Delta V$$


* **Práce vykonaná na výstupu (bod 2):** Systém musí konat práci, aby vytlačil tekutinu proti vnějšímu tlaku $p_2 A_2$. Práce je záporná:

$$W_2 = -F_2 \Delta s_2 = -p_2 A_2 \Delta s_2 = -p_2 \Delta V$$


* **Celková vykonaná práce ($W$):**

$$W = W_1 + W_2 = (p_1 - p_2) \Delta V$$

Dále sledujeme změnu energie hmotnosti $\Delta m = \rho \Delta V$, která se efektivně přemístila z výšky $z_1$ a rychlosti $v_1$ do výšky $z_2$ a rychlosti $v_2$.

* **Změna kinetické energie ($\Delta E_k$):**

$$\Delta E_k = \frac{1}{2} \Delta m v_2^2 - \frac{1}{2} \Delta m v_1^2 = \frac{1}{2} \rho \Delta V (v_2^2 - v_1^2)$$


* **Změna potenciální energie ($\Delta E_p$):**

$$\Delta E_p = \Delta m g z_2 - \Delta m g z_1 = \rho \Delta V g (z_2 - z_1)$$


Podle zákona zachování energie musí platit, že práce vykonaná na systému se rovná změně jeho energie:

$$W = \Delta E_k + \Delta E_p$$

Dosadíme předchozí vztahy:

$$(p_1 - p_2) \Delta V = \frac{1}{2} \rho \Delta V (v_2^2 - v_1^2) + \rho \Delta V g (z_2 - z_1)$$

Vydělíme celou rovnici objemem $\Delta V$ (čímž získáme energie vztažené na jednotku objemu):

$$p_1 - p_2 = \frac{1}{2} \rho v_2^2 - \frac{1}{2} \rho v_1^2 + \rho g z_2 - \rho g z_1$$


Přeskupením členů tak, aby indexy 1 byly na jedné straně a indexy 2 na druhé, dostáváme finální tvar:

$$p_1 + \frac{1}{2} \rho v_1^2 + \rho g z_1 = p_2 + \frac{1}{2} \rho v_2^2 + \rho g z_2 = \text{konst.}$$

**Fyzikální význam členů:**

* $p$: Hustota tlakové energie (statický tlak).
* $\frac{1}{2} \rho v^2$: Hustota kinetické energie (dynamický tlak).
* $\rho g z$: Hustota potenciální energie (hydrostatický tlak).

Tato rovnice předpokládá, že do systému není dodáváno teplo ani mechanická práce (např. čerpadlem).


### Nevířivé (irotační) proudění

Závěrem si definujme speciální, ale v aerodynamice velmi důležitou kinematickou vlastnost. Proudění nazýváme nevířivým (irotačním) tehdy, když absolutně žádná částice tekutiny nevykonává rotační pohyb kolem své vlastní osy.

Matematicky to popisujeme tak, že vektorová rotace (v anglosaské literatuře *curl*) rychlostního pole je v každém bodě nulová:

$$ \nabla \times \vec{v} = 0 $$

Pokud proudění omezíme do 2D roviny (osy $x, z$), nevířivost se redukuje na jednoduchou podmínku rovnosti křížových parciálních derivací složek rychlosti:

$$ \frac{\partial u}{\partial z} = \frac{\partial w}{\partial x} $$.
