##  Charakteristika proudění a viskozita

Při toku tekutiny se mohou její různé vrstvy pohybovat vůči sobě odlišnými rychlostmi, přičemž jedna vrstva klouže po druhé. Abychom tomuto pohybu porozuměli, musíme rozlišovat dva základní režimy proudění a pochopit síly, které v tekutině působí.

Pokud se podíváme na tekutinu proudící v dlouhém válcovém potrubí při nízkých rychlostech, částice tekutiny se pohybují po drahách rovnoběžných se stěnou potrubí. Dostatečně daleko od vstupu do potrubí se proudění stává ustáleným (plně vyvinutým).

Toto ustálené, uspořádané proudění ve vrstvách se nazývá **laminární proudění**.

Základním experimentálně ověřeným faktem pro všechny body, kde je tekutina v kontaktu se stěnou, je tzv. **podmínka ulpívání (no-slip condition)** – tekutina u stěny potrubí je vůči stěně v klidu (její rychlost je nulová). Rychlost tekutiny se směrem do nitra potrubí zvyšuje a svého maxima ($v_{max}$) dosahuje přesně ve středu.

```{note}
Tento parabolický rychlostní profil má nenulový gradient rychlosti ve směru kolmém na směr proudění.

```

Kvůli válcové geometrii potrubí po sobě jednotlivé válcové vrstvy tekutiny kloužou, což vede ke vzniku tečných sil mezi vrstvami. Tečná síla působící na jednotku plochy se nazývá **smykové napětí** ($\tau$).

Mírou odporu proti tomuto vzájemnému klouzání vrstev je **viskozita** tekutiny. Ideální (dokonalá) tekutina nemá mezi vrstvami žádné tečné síly. Reálnou tekutinu nazýváme **newtonovskou**, pokud jsou smykové síly na jednotku plochy přímo úměrné gradientu rychlosti.

Pro newtonovskou tekutinu v laminárním toku ve válcovém potrubí je smykové napětí dáno vztahem:

$$\tau = \eta \frac{dv}{dr}$$

Kde $\eta$ je konstanta úměrnosti nazývaná absolutní (dynamická) viskozita, $r$ je radiální vzdálenost od středové osy potrubí a $\frac{dv}{dr}$ je gradient rychlosti kolmý na směr proudění. Jednotkou viskozity v soustavě SI je $\text{Pa}\cdot\text{s}$ (neboli $\text{N}\cdot\text{s}/\text{m}^2$).


### Odvození rychlostních profilů v laminárním toku

#### Couetteovo proudění (Proudění mezi deskami)

Uvažujme newtonovskou tekutinu mezi dvěma velmi dlouhými paralelními deskami o šířce $w$, délce $l$, které jsou od sebe vzdáleny o $h$. Horní deska se pohybuje konstantní relativní rychlostí $v_0$ vůči spodní desce, která je v klidu (v rovině $y = 0$).

Vyberme si objemový element o délce $\Delta x$ a průřezu $A$. Smyková síla na tento element od tekutiny nad ním je:


$$F_{smyk} = \tau A = \eta A \frac{dv}{dy}$$

Tato síla je vyrovnávána smykovou silou spodní desky na element, takže síly jsou v rovnováze a smykové napětí $\tau$ je konstantní. Rychlost tekutiny na spodní desce je nulová. Integrací diferenciální rovnice získáme:

$$\int_0^v dv = \int_0^y \frac{\tau}{\eta} dy$$

$$v(y) = \frac{\tau}{\eta} y$$

Protože rychlost na horní desce (v $y = h$) je $v_0$, platí pro konstantní smykové napětí $\tau = \frac{\eta v_0}{h}$. Rychlostní profil je tedy lineární:

$$v(y) = \frac{v_0}{h} y$$

Tento typ toku je známý jako **Couetteovo proudění**.

#### Hagen-Poiseuilleovo proudění (Laminární tok v potrubí)

Nyní uvažujme dlouhé válcové potrubí o poloměru $R$, kde se částice tekutiny pohybují paralelně s osou. Zvolme válcový objemový element o délce $l$ a poloměru $r$ centrovaný na ose potrubí.

Na koncích tohoto elementu je tlakový spád $\Delta p$. Součet sil na objemový element je nulový a skládá se z tlakové síly na podstavy (tlakový rozdíl $\times$ plocha) a smykové síly na plášť válce:

$$\Delta p (\pi r^2) + \tau (2\pi r l) = 0$$

S využitím našeho newtonovského modelu $\tau = \eta \frac{dv}{dr}$ dostáváme:

$$\Delta p (\pi r^2) + \left( \eta \frac{dv}{dr} \right) 2\pi r l = 0$$

Tuto rovnici upravíme pro separaci proměnných s okrajovou podmínkou podmínky ulpívání ($v(R) = 0$):

$$dv = -\frac{\Delta p}{2\eta l} r \, dr$$

Integrací od $r$ do $R$ získáme rovnici paraboly:

$$\int_{v(r)}^{0} dv' = -\int_{r}^{R} \frac{\Delta p}{2\eta l} r' \, dr'$$

$$-v(r) = -\frac{\Delta p}{4\eta l} (R^2 - r^2) \implies v(r) = \frac{\Delta p}{4\eta l} (R^2 - r^2)$$

Maximální rychlost uprostřed potrubí ($r=0$) je:

$$v_{max} = \frac{\Delta p R^2}{4\eta l}$$

### Objemový průtok a průměrná rychlost

Objemový průtok ($Q$) získáme integrací rychlostního profilu přes plochu prstenců průřezu potrubí ($dA = 2\pi r dr$):

$$Q = \int_0^R v(r) 2\pi r \, dr = \int_0^R \frac{\Delta p}{4\eta l} (R^2 - r^2) 2\pi r \, dr = \frac{\pi \Delta p R^4}{8\eta l}$$

Průměrná rychlost toku je pak dána celkovým průtokem děleným plochou průřezu ($\pi R^2$):

$$v_{avg} = \frac{\Delta p R^2}{8\eta l}$$

Z toho jasně vyplývá, že průměrná rychlost je přesně polovinou maximální rychlosti ve středu potrubí ($v_{avg} = v_{max}/2$). Rovnici lze přepsat pomocí průměru potrubí $D = 2R$:

$$\Delta p = \frac{32 \eta l v_{avg}}{D^2}$$


## Reynoldsovo číslo a turbulentní proudění

Při dosažení určitého průtoku se odpor tekutiny náhle zvýší a částice přestanou sledovat rovné linie. Začnou se pohybovat náhodně (chaoticky), i když průměrný pohyb stále směřuje podél osy potrubí. Tento typ proudění nazýváme **turbulentní proudění**. Bylo experimentálně zjištěno, že i malá porucha v laminárním toku může rychle narůst a vyvolat turbulenci.

Osbourne Reynolds byl první, kdo experimentálně tyto dva režimy změřil a dokázal charakterizovat bod přechodu pomocí bezrozměrného parametru, který závisí na průměrné rychlosti, průměru potrubí a viskozitě tekutiny. Tento parametr se nazývá **Reynoldsovo číslo ($Re$)** a je definován jako poměr setrvačných a viskózních sil v tekutině:

$$Re = \frac{\rho v_{avg} D}{\eta}$$

Přechodový bod mezi laminárním a turbulentním prouděním odpovídá kritické hodnotě Reynoldsova čísla, při které dochází k náhlému nárůstu tření (odporu) mezi vrstvami tekutiny.

## Ztrátový součinitel tření

Odpor potrubí (tlaková ztráta) se v inženýrství často vyjadřuje pomocí měrné tlakové ztráty na jednotku délky potrubí ($\Delta p / l$). Lze ji také vyjádřit pomocí ztrátového součinitele $k$:

$$\frac{\Delta p}{l} = k \frac{1}{2} \rho v_{avg}^2$$

Pro dlouhé přímé válcové potrubí je ztrátový součinitel $k$ zapsán pomocí součinitele tření (friction factor) $f$ a ekvivalentní délky potrubí $1/D$:

$$k = f \frac{1}{D}$$

Dosazením získáme takzvanou Darcy-Weisbachovu rovnici:

$$\Delta p = f \frac{l}{D} \frac{1}{2} \rho v_{avg}^2$$

Pokud srovnáme analyticky odvozenou tlakovou ztrátu pro laminární tok ($\Delta p = \frac{32 \eta l v_{avg}}{D^2}$) s empirickou Darcy-Weisbachovou rovnicí, dostaneme vztah pro určení součinitele tření $f$:

$$\frac{32 \eta l v_{avg}}{D^2} = f \frac{l}{D} \frac{1}{2} \rho v_{avg}^2$$

$$f = \frac{64 \eta}{\rho v_{avg} D}$$

Protože jsme si definovali Reynoldsovo číslo jako $Re = \frac{\rho v_{avg} D}{\eta}$, můžeme ztrátový součinitel pro laminární proudění vyjádřit elegantně jako:

$$f = \frac{64}{Re}$$

V turbulentním režimu se mechanismus odporu radikálně mění. Náhodný pohyb částic a vznik vírů způsobuje rozptyl kinetické energie (disipaci) do mnohem větší míry, než je tomu u viskózního tření hladkých vrstev. Ztrátový součinitel $f$ v turbulentním toku proto klesá s Reynoldsovým číslem mnohem pomaleji (není již nepřímo úměrný $1/Re$) a začíná být silně závislý na drsnosti stěn potrubí. Výpočet odporové síly u turbulencí tak opouští roviny jednoduché diferenciální analýzy a opírá se převážně o empirické vztahy (např. Colebrookova rovnice) a Moodyho diagram.
