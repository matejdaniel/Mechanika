## Odporová síla obtékaného tělesa (Drag Force)

Když se pevné těleso pohybuje tekutinou (nebo tekutina obtéká pevné těleso), tekutina na něj vyvíjí celkovou aerodynamickou (či hydrodynamickou) sílu. Složka této síly, která působí **paralelně se směrem proudu a proti směru pohybu tělesa**, se nazývá **odporová síla** ($F_D$, z anglického *Drag force*).

Celkový odpor tělesa se skládá ze dvou zcela odlišných fyzikálních fenoménů:

1. **Třecí odpor (Skin friction drag):** Vzniká v důsledku viskozity tekutiny a podmínky ulpívání v tzv. mezní vrstvě těsně u povrchu tělesa. Jde o integrál tečných (smykových) napětí $\tau$ přes celý povrch.
2. **Tvarový odpor (Form drag / Pressure drag):** Vzniká v důsledku rozdílu tlaků mezi přední (návětrnou) a zadní (závětrnou) stranou tělesa. Pokud se proud na zadní straně odtrhne od povrchu tělesa, vznikne oblast s nízkým tlakem (úplav), která těleso "nasává" dozadu.

Pro inženýrské výpočty byla na základě rozměrové analýzy a experimentů odvozena univerzální Newtonova rovnice pro odporovou sílu:

$$F_D = \frac{1}{2} C_D \rho A v^2$$

Kde:

* $F_D$ je odporová síla (N).
* $\rho$ je hustota tekutiny (kg/m³).
* $v$ je rychlost proudu vůči tělesu (m/s).
* $A$ je referenční plocha tělesa (m²). U tupých těles se obvykle bere plocha čelního průmětu (stín, který těleso vrhá ve směru proudu), u křídel se bere půdorysná plocha.
* $C_D$ je **součinitel odporu** (bezrozměrný). Závisí na tvaru tělesa a na Reynoldsově čísle ($Re$).

Pojďme se nyní podívat, jak se tento odpor chová v čistě laminárním a v turbulentním režimu.

### Odporová síla v laminárním proudění (Plíživé proudění)

Představme si makroskopické těleso obtékané velmi pomalu, tekutinou o vysoké viskozitě (např. kulička padající v medu) nebo těleso mikroskopických rozměrů (např. bakterie či červená krvinka v plazmě). V těchto případech je Reynoldsovo číslo extrémně malé ($Re \ll 1$). Setrvačné síly jsou zanedbatelné a obtékání dominují viskózní síly.

Proudění je **dokonale laminární**, proudnice se úhledně rozestoupí před tělesem a plynule se zase spojí za ním. Nedochází k odtržení proudu, takže **tvarový (tlakový) odpor je téměř nulový**. Dominantní je třecí odpor.

### Stokesův zákon pro kouli

George Gabriel Stokes analyticky vyřešil Navier-Stokesovy rovnice pro tento specifický případ plíživého (laminárního) obtékání dokonalé koule o poloměru $R$. Zjistil, že odporová síla je v tomto režimu **přímo úměrná rychlosti** (nikoliv jejímu kvadrátu!):

$$F_D = 6 \pi \eta R v$$

Kde $\eta$ je dynamická viskozita.

```{admonition} Odvození součinitele odporu pro Stokesův tok
:class: hint
Pokud Stokesovu rovnici dosadíme do obecné definice odporové síly ($F_D = \frac{1}{2} C_D \rho A v^2$) a jako referenční plochu koule dosadíme $A = \pi R^2$, můžeme vyjádřit součinitel odporu $C_D$:

$$6 \pi \eta R v = \frac{1}{2} C_D \rho (\pi R^2) v^2$$

Algebraickou úpravou a vykrácením dostaneme:

$$C_D = \frac{12 \eta}{\rho R v}$$

Reynoldsovo číslo pro kouli je definováno přes její průměr $D = 2R$, tedy $Re = \frac{\rho v (2R)}{\eta}$. Po dosazení $Re$ do vztahu pro $C_D$ získáme slavný vztah pro součinitel odporu v laminárním režimu:

$$C_D = \frac{24}{Re}$$

```

Z tohoto odvození plyne, že v čistě laminárním (viskózním) režimu je součinitel odporu silně závislý na rychlosti (klesá s rostoucím $Re$).

---

## 4. Odporová síla v turbulentním proudění (Setrvačný režim)

Jak rychlost roste, Reynoldsovo číslo se zvyšuje. V mezní vrstvě tekutiny u povrchu tělesa dochází ke ztrátě hybnosti kvůli tření. Tekutina pak už nemá dostatek kinetické energie na to, aby překonala rostoucí tlak na zadní straně profilu.

Výsledkem je **odtržení mezní vrstvy**. Proudnice již nekopírují tvar tělesa. Za tělesem vzniká chaotická, vířivá oblast zvaná **úplav (wake)**.

V tomto úplavu je výrazně nižší tlak než na přední straně tělesa. Tento obrovský tlakový rozdíl způsobuje, že **tvarový odpor zcela převládne nad třecím odporem**.

### Kvadratická závislost

Při vysokých Reynoldsových číslech ($Re > 1000$), kdy je proudění v úplavu plně turbulentní, se součinitel odporu $C_D$ ustaluje a stává se téměř konstantním (nezávislým na Reynoldsově čísle, závisí už jen na samotném tvaru profilu).

Pokud je $C_D$ konstantní, obecná rovnice nám říká, že:

$$F_D \propto v^2$$

Odporová síla v turbulentním režimu (což je případ jedoucího auta, letícího míčku nebo letadla) roste s **druhou mocninou rychlosti**. Pokud auto zrychlí na dvojnásobek, musí motor překonávat čtyřnásobný aerodynamický odpor.

```{important} Krize odporu (Tajemství golfového míčku)
U tupých těles existuje fascinující anomálie zvaná *krize odporu*. Pokud se hladká mezní vrstva na přední straně tělesa stane sama o sobě turbulentní (například při dosažení $Re \approx 10^5$), turbulentní víry v této vrstvě přinesou z okolí novou kinetickou energii těsně k povrchu.

Tato "doplněná" energie způsobí, že tekutina dokáže vzdorovat odtržení mnohem déle. Proud se odtrhne až více vzadu, čímž se zadní úplav (zóna nízkého tlaku) dramaticky zmenší. Ačkoliv se mírně zvýší třecí odpor, **tlakový odpor klesne tak prudce, že celková odporová síla $F_D$ tělesa klesne**. 

To je důvod, proč mají golfové míčky na povrchu důlky – záměrně vyvolávají turbulenci v mezní vrstvě, zmenšují zadní úplav a míček díky menšímu odporu doletí mnohem dále.

```
