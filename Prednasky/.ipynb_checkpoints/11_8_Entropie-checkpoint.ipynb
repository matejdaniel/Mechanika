## Entropie

V předchozí kapitole jsme se dozvěděli, že druhý termodynamický zákon určuje směr dějů a limity jejich účinnosti. Nyní představíme způsob, jak tyto koncepty převést do matematické a analytické podoby, kterou inženýři potřebují k reálným výpočtům. Stejně jako je první zákon spojen s veličinou *energie*, druhý zákon je bytostně spjat s veličinou zvanou **entropie**.

Ačkoliv pojem energie znáte z běžného života, pojem entropie slyšíte pravděpodobně poprvé. Obě tyto vlastnosti jsou abstraktní, ale pro termodynamickou analýzu naprosto klíčové. K entropii se nedobereme tak, že si představíme nějakou "látku" uvnitř systému, ale matematickou úvahou vycházející z druhého termodynamického zákona.

### Clausiova nerovnost

Celý příběh entropie začíná matematickým důsledkem Kelvin-Planckovy formulace druhého zákona, který odvodil Rudolf Clausius a který se nazývá **Clausiova nerovnost**. Ta platí pro absolutně jakýkoliv termodynamický cyklus (vratný i nevratný) a říká:

$$ \oint \left( \frac{\delta Q}{T} \right)_b \le 0 $$

Co tento integrál přes uzavřenou křivku cyklu ($\oint$) znamená?

* $\delta Q$ je teplo přijaté nebo odevzdané systémem v daném okamžiku.
* $T$ je absolutní teplota (v Kelvinech) na hranici systému ($b$ - boundary) přesně v místě a čase, kde k přenosu tepla dochází.
* My postupně sčítáme tyto zlomky ($\delta Q/T$) po celou dobu trvání cyklu, dokud se nevrátíme do výchozího bodu.

Clausiova nerovnost tvrdí, že tento součet nikdy nemůže být kladný. Má pouze dvě možnosti:

1. **Je přesně roven nule:** To platí pouze v případě, že cyklus je **vnitřně vratný** (uvnitř systému nedochází k žádnému tření, náhlým expanzím ani jiným nevratnostem).
2. **Je menší než nula (záporný):** To platí pro všechny reálné cykly, kde se vyskytují vnitřní nevratnosti.

Abychom z nerovnosti udělali rovnici a mohli vyjádřit "míru" těchto nevratností, můžeme zavést veličinu $\sigma_{cycle}$, která je vždy kladná (nebo nulová), a rovnici přepsat:

$$ \oint \left( \frac{\delta Q}{T} \right)_b = -\sigma_{cycle} $$

Čím více vnitřních nevratností v cyklu je (tření atd.), tím větší bude hodnota $\sigma_{cycle}$ a tím zápornější bude hodnota integrálu. Pokud je cyklus vnitřně vratný, $\sigma_{cycle} = 0$.

### Definice entropie (Změna entropie)

Nyní provedeme myšlenkový experiment s vnitřně vratnými procesy. Představme si cyklus, který začíná ve stavu 1, do stavu 2 se dostane vnitřně vratným procesem $A$ a zpět do stavu 1 se vrátí jiným vnitřně vratným procesem $C$. Protože je celý cyklus vratný, Clausiova rovnice říká, že integrál za celý cyklus je nula:

$$ \int_{1}^{2} \left( \frac{\delta Q}{T} \right)_A + \int_{2}^{1} \left( \frac{\delta Q}{T} \right)_C = 0 $$

Co když ale ze stavu 1 do stavu 2 použijeme úplně jiný vnitřně vratný proces, označme ho $B$, a vrátíme se stejnou cestou $C$? Opět musí platit:

$$ \int_{1}^{2} \left( \frac{\delta Q}{T} \right)_B + \int_{2}^{1} \left( \frac{\delta Q}{T} \right)_C = 0 $$

Pokud od sebe tyto dvě rovnice odečteme, integrály pro cestu $C$ se vyruší a zjistíme neuvěřitelnou věc:

$$ \int_{1}^{2} \left( \frac{\delta Q}{T} \right)_A = \int_{1}^{2} \left( \frac{\delta Q}{T} \right)_B $$

Tento výsledek znamená, že **hodnota integrálu $\int \frac{\delta Q}{T}$ vypočtená pro vnitřně vratný děj nezávisí na tom, jakou cestou jsme se ze stavu 1 do stavu 2 dostali**. Záleží pouze na počátečním a konečném stavu!

Z první kapitoly víme naprosto jistě: pokud veličina závisí pouze na počátečním a konečném stavu a vůbec nezávisí na procesu, pak **tato veličina musí být vlastností systému (stavovou veličinou)**.

Tuto nově objevenou vlastnost systému označujeme symbolem **$S$** a nazýváme ji **Entropie**. Změna entropie mezi dvěma stavy je tedy definována jako:

$$ S_2 - S_1 = \int_{1}^{2} \left( \frac{\delta Q}{T} \right)_{\text{int rev}} $$

*Poznámka: Index "int rev" zdůrazňuje, že ačkoliv reálný proces může být nevratný, změnu entropie mezi jeho koncovými stavy musíme matematicky vypočítat tak, že si mezi nimi představíme jakýkoliv (třeba i vymyšlený) vnitřně vratný proces a zintegrujeme pro něj teplo lomeno teplotou.*

V diferenciálním tvaru píšeme definiční rovnici jako:


$$ dS = \left( \frac{\delta Q}{T} \right)_{\text{int rev}} $$

### Vlastnosti entropie

* Entropie je **extenzivní vlastnost** (čím větší systém, tím větší entropie).
* Značí se velkým $S$ (pro celý systém) nebo malým $s$ jako měrná entropie vztažená na 1 kg hmotnosti.
* Základní jednotkou v SI je **J/K**, pro inženýrské účely používáme měrnou entropii v **kJ/(kg·K)**.
* Změna entropie je přesně dána pouze počátečním a konečným stavem bez ohledu na to, zda k této změně došlo vratně, nebo s masivním třením a nevratnostmi. Entropie je zkrátka parametr, který charakterizuje stav látky úplně stejně jako tlak nebo teplota.

Podobně jako u entalpie, definice entropie vznikla z matematického odvození integrálů. Fyzikálně si ji lze představit (jak jsme ukázali ve statistické mechanice) jako míru neuspořádanosti systému, nebo inženýrsky: jako nástroj, který nám umožní bilancovat kvalitu energie a počítat účinnosti turbín a kompresorů, jak uvidíme v další podkapitole.

### Změny entropie v Carnotově cyklu a T-s diagram

V minulé kapitole jsme si definovali změnu entropie pro vnitřně vratný děj jako integrál tepla lomeno teplotou:

$$ dS = \left( \frac{\delta Q}{T} \right)_{\text{int rev}} $$

Zároveň víme, že **Carnotův cyklus** se skládá přesně ze čtyř vnitřně vratných dějů. Pojďme se tedy podívat, co se děje s entropií v každém z těchto čtyř kroků.

### Analýza jednotlivých dějů

**1. Vratná izotermická expanze (1 $\to$ 2)**
Systém je v kontaktu s horkou tepelnou lázní o stálé teplotě $T_H$ a přijímá od ní teplo $Q_H$. Protože je teplota $T_H$ konstantní, můžeme ji z integrálu vytknout:


$$ \Delta S_{12} = S_2 - S_1 = \int_{1}^{2} \frac{\delta Q}{T_H} = \frac{1}{T_H} \int_{1}^{2} \delta Q = \frac{Q_H}{T_H} $$


*Výsledek:* Systém přijal teplo, jeho entropie **vzrostla** o hodnotu $Q_H/T_H$.

**2. Vratná adiabatická expanze (2 $\to$ 3)**
Systém je tepelně izolován, nedochází k žádnému přenosu tepla ($\delta Q = 0$). Plyn expanduje a jeho teplota klesne z $T_H$ na $T_C$.


$$ \Delta S_{23} = S_3 - S_2 = \int_{2}^{3} \frac{0}{T} = 0 $$


*Výsledek:* Entropie se **nemění** ($S_3 = S_2$). Děj, při kterém zůstává entropie konstantní, se nazývá **izentropický děj**. *(Každý vratný adiabatický děj je automaticky izentropický).*

**3. Vratná izotermická komprese (3 $\to$ 4)**
Systém je v kontaktu s chladnou tepelnou lázní o stálé teplotě $T_C$ a odevzdává jí teplo $Q_C$. (Z termodynamického hlediska je teplo odevzdané záporné).


$$ \Delta S_{34} = S_4 - S_3 = \int_{3}^{4} \frac{\delta Q}{T_C} = -\frac{Q_C}{T_C} $$


*Výsledek:* Systém odevzdal teplo, jeho entropie **klesla** o hodnotu $Q_C/T_C$.

**4. Vratná adiabatická komprese (4 $\to$ 1)**
Systém je opět izolován ($\delta Q = 0$), stlačujeme ho a jeho teplota roste zpět na $T_H$.


$$ \Delta S_{41} = S_1 - S_4 = \int_{4}^{1} \frac{0}{T} = 0 $$


*Výsledek:* Entropie se **nemění** ($S_1 = S_4$). Tento děj je opět izentropický.


####  Teplotně-entropický (T-s) diagram

V kapitole o plynech jsme si ukazovali pracovní (p-v) diagram, kde plocha pod křivkou děje znamenala objemovou práci ($W = \int p \, dv$).

Z definice entropie ($dS = \delta Q / T$) si ale můžeme matematicky vyjádřit teplo:


$$ \delta Q_{\text{int rev}} = T \, dS \implies Q_{\text{int rev}} = \int T \, dS $$

To znamená, že pokud vyneseme **teplotu na osu Y** a **entropii na osu X**, získáme tzv. **T-s diagram**, kde **plocha pod křivkou děje odpovídá vyměněnému teplu**.

Tento diagram je pro termodynamiku naprosto vizionářský. Pokud si do něj totiž zakreslíme náš Carnotův cyklus, zjistíme úžasnou věc:

* Izotermické děje (stálá teplota) jsou **vodorovné úsečky**.
* Izentropické děje (stálá entropie) jsou **svislé úsečky**.

Carnotův cyklus tak v T-s diagramu tvoří naprosto **dokonalý obdélník**! Plocha uvnitř tohoto obdélníku představuje rozdíl mezi přijatým a odevzdaným teplem ($Q_H - Q_C$), což se podle první věty termodynamické rovná **čisté užitečné práci cyklu ($W_{net}$)**.

Z obdélníku v T-s diagramu lze navíc okamžitě, pouze ze znalosti obsahu obdélníka, odvodit Carnotovu účinnost:

* Dodané teplo (plocha pod horní izotermou až k ose X): $Q_H = T_H \cdot \Delta S$
* Užitečná práce (plocha samotného obdélníku): $W_{net} = (T_H - T_C) \cdot \Delta S$
* Účinnost: $\eta = \frac{W_{net}}{Q_H} = \frac{(T_H - T_C)\Delta S}{T_H \Delta S} = \frac{T_H - T_C}{T_H} = 1 - \frac{T_C}{T_H}$

Entropie tak dává Carnotově účinnosti krásný, jednoduchý geometrický smysl.
