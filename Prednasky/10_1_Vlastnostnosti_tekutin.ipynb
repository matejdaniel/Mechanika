## Fyzikální vlastnosti tekutin

### Hustota (Density, $\rho$)

**Hustota** je definována jako hmotnost látky připadající na jednotku objemu.

*   **Vliv prostředí:** Hustota se může měnit v závislosti na tlaku a teplotě (např. při zahřívání vzduchu se jeho hustota snižuje a vzduch stoupá).
*   **Referenční hodnoty:**
    *   Hustota sladké vody při **4 °C** je **1000 kg/m³**; při **20 °C** mírně klesne na **998,2 kg/m³**.
    *   Hustota mořské vody je přibližně **1025 kg/m³** (při **20 °C** a tlaku 1 atmosféry).
    *   Hustota vzduchu je **1,204 kg/m³** (při **20 °C** a atmosférickém tlaku). 
    *   *Poznámka:* Voda je tedy zhruba tisíckrát hustší než vzduch!
*   **Stlačitelnost:** Pokud se hustota tekutiny při změně tlaku nemění, považujeme tekutinu za **nestlačitelnou**. Stlačitelnost úzce souvisí s rychlostí zvuku v daném médiu (čím hustší tekutina, tím vyšší rychlost zvuku – např. ve vzduchu **300 m/s**, ve vodě cca **1000 m/s**). Tekutina se začne projevovat jako stlačitelná teprve tehdy, když rychlost jejího proudění překročí rychlost zvuku (což vede ke studiu subsonického, transonického a supersonického proudění v aerodynamice). Pro většinu aplikací spojených s vodou (a pro účely tohoto kurzu) budeme předpokládat, že je tekutina nestlačitelná.


### Relativní hustota (Specific Gravity, SG)

**Relativní hustota** je bezrozměrný poměr hustoty dané látky k hustotě referenční látky (nejčastěji vody).

*   Pokud je $SG < 1$, látka bude ve vodě plavat.
*   Příklad: Led plave na vodě, protože jeho relativní hustota je přibližně **0,917** (při **20 °C**).

### Tlak (Pressure, $p$)

**Tlak** je skalární veličina, která vždy působí kolmo (v pravých úhlech) na daný povrch.

*   **Izotropie:** Tlak je v daném bodě izotropní, což znamená, že je stejný ve všech směrech a působí bez ohledu na orientaci plochy.
*   **Síla a napětí:** Tlak je v podstatě normálové napětí (síla na jednotku plochy). I nepatrný tlak dokáže vyvolat obrovskou sílu, pokud působí na dostatečně velkou plochu.
    *   *Příklad s balónkem:* Tlak vzduchu uvnitř rovnoměrně tlačí na stěny balónku směrem ven. Pokud tlak příliš vzroste, stěny se ztenčí a balónek praskne.
    *   *Příklad s potrubím:* Tlak uvnitř trubky působí radiálně ven, což vyvolává tahové napětí ve stěnách potrubí.
*   **Přenos tlaku:** Tlak se tekutinou šíří rovnoměrně. Zvýšíte-li tlak v jednom bodě (např. zatlačením prstu do balónku), tato změna se rozšíří celým plynem. Tyto tlakové změny se šíří rychlostí zvuku.
*   **Dynamický tlak:** Zvýšení rychlosti proudění ovlivňuje tlak. Pokud se částice tekutiny urychlí z rychlosti $V_1$ na $V_2$ (ve stejné výšce), změna tlaku (vyplývající z Bernoulliho rovnice) bude:
    $$ \Delta p = -\frac{1}{2} \rho (v_2^2 - v_1^2) $$

### Viskozita (Viscosity)

Viskozita vyjadřuje ochotu tekutiny volně téct a určuje, jak tekutina reaguje na mechanické napětí a deformaci (představuje míru "lepkavosti" nebo vnitřního tření tekutiny). Různé tekutiny tečou různě snadno (např. voda vs. motorový olej).



*   **Smykové napětí a rychlost deformace:** U tzv. newtonovských tekutin (voda, olej, vzduch) existuje lineární vztah mezi smykovým napětím ($\tau$) a rychlostí deformace. Rychlost deformace lze geometricky zjednodušit na gradient rychlosti ($\frac{du}{dy}$).

Představme si malý element tekutiny. Pokud se vrstvy tekutiny pohybují různými rychlostmi, dochází k jeho deformaci (zkosení). 

Z geometrie této deformace vyplývá vztah pro úhel zkosení $\delta\theta$:

$$ \tan \delta\theta = \frac{\delta v \delta t}{\delta y} $$

*(Vysvětlení: Vzdálenost, o kterou se horní vrstva posune vůči spodní za čas $\delta t$, je dána rozdílem rychlostí $\delta v$ vynásobeným časem $\delta t$. Tuto vzdálenost pak dělíme tloušťkou vrstvy $\delta y$, čímž získáme tangens úhlu deformace).*

Pokud se omezíme na infinitezimálně malé změny (velmi malé úhly), můžeme využít matematickou aproximaci pro malé úhly: $\tan \delta\theta \approx \delta\theta$. Rovnici pak můžeme upravit vydělením časem $\delta t$ a přechodem k diferenciálům:

$$ \frac{d\theta}{dt} = \frac{dv}{dy} $$

Slovy to znamená zásadní fyzikální fakt pro tekutiny:

**Rychlost smykové deformace = gradient rychlosti**

Jelikož je smykové napětí ($\tau$) lineárně úměrné rychlosti smykové deformace, musí být logicky úměrné i gradientu rychlosti. V hydrodynamice se ukázalo, že koncept "rychlosti deformace" ($\frac{d\theta}{dt}$) není z hlediska měření a výpočtů tak praktický jako **gradient rychlosti v tekutině** ($\frac{du}{dy}$), protože rychlost proudění lze snadněji měřit a modelovat.

Proto se vztah úměrnosti pro smykové napětí primárně zapisuje právě pomocí gradientu rychlosti. Výsledkem je slavný Newtonův zákon viskozity:

$$ \tau = \eta \frac{d\theta}{dt} = \eta \frac{du}{dy} $$


*   **Dynamická viskozita ($\eta$):** Vyjadřuje konstantu úměrnosti mezi smykovým napětím a gradientem rychlosti:

    $$ \tau = \eta \frac{du}{dy} $$

Její jednotkou je **N·s/m²** (neboli **kg/(m·s)**).

*   **Kinematická viskozita ($\nu$):** Je poměrem dynamické viskozity a hustoty tekutiny. Často se používá, protože přímo souvisí s pohybem (kinematikou) tekutiny:

$$ \nu = \frac{\mu}{\rho} $$

Jednotkou kinematické viskozity je **m²/s**.

*   **Teplotní závislost:** Obě viskozity se mění s teplotou. Pro vzduch při **20 °C** je $\mu \approx$ **1,82 × 10⁻⁵ N·s/m²**. Pro vodu při **20 °C** je to přibližně **1,0 × 10⁻³ N·s/m²**.
*   **Podmínka ulpívání (No-slip condition):** Kvůli viskozitě tekutina "ulpívá" na pevných površích. To znamená, že rychlost tekutiny těsně u stěny tělesa je nulová. Tento jev vede ke vzniku tzv. **mezní vrstvy** (boundary layer), což je tenká vrstva tekutiny u povrchu, kde se rychlost postupně zvyšuje z nuly na rychlost volného proudu.

### Bezrozměrné parametry a Reynoldsovo číslo (Reynolds Number, $Re$)

Bezrozměrné parametry jsou v mechanice tekutin klíčové, protože umožňují porovnávat a škálovat experimenty (např. testování zmenšeného modelu lodi a aplikování výsledků na skutečnou loď).

**Reynoldsovo číslo** je jedním z nejdůležitějších parametrů. Je definováno jako poměr setrvačných sil k silám viskózním:

$$ Re = \frac{U L}{\nu} $$

(Kde $U$ je charakteristická rychlost proudění, $L$ je charakteristický délkový rozměr a $\nu$ je kinematická viskozita). Reynoldsovo číslo mimo jiné indikuje, jak silná bude mezní vrstva a zda bude proudění laminární či turbulentní.

### Povrchové napětí (Surface Tension, $\sigma$)

**Povrchové napětí** hraje významnou roli na rozhraní mezi kapalinou a plynem (tedy v místech s nespojitostí hustoty). 

*   **Příčina:** Vzniká v důsledku přitažlivých (kohezních) sil mezi molekulami. Molekuly uvnitř kapaliny jsou obklopeny ze všech stran a síly se vyrovnávají. Molekuly na hladině však nemají sousedy směrem vzhůru, což vytváří výsledné tahové napětí podél povrchu.
*   **Důsledek:** Kapaliny mají tendenci stahovat svůj povrch na co nejmenší možnou plochu (to je důvod, proč vodní kapky tvoří kulový tvar).
*   **Jednotky:** Povrchové napětí se udává jako práce potřebná k vytvoření jednotkové plochy, tedy v jednotkách **N/m** (Newton na metr).
