## Síly v tekutinách

Síly v tekutinách mohou vznikat v důsledku napětí v proudící tekutině (tlak a viskózní smykové napětí), 
gravitačních sil, zrychlení tekutiny nebo jiných objemových sil. Prozatím budeme uvažovat **tekutinu ve statické rovnováze** – tedy stav, kdy v tekutině neexistují žádné rychlostní gradienty 
(a tím pádem ani žádná viskózní smyková napětí, protože tekutina neproudí).

Síly působící na tekutinu ve statické rovnováze jsou způsobeny **pouze**:
1.  **Tlakem** působícím na objem tekutiny.
2.  **Gravitací** působící na hmotnost tekutiny.
3.  **Dalšími vnějšími objemovými silami** (např. elektromagnetickými – ty ale v běžné hydrodynamice většinou zanedbáváme).

Podle standardních konvencí považujeme tlak za kladný, pokud způsobuje stlačení (kompresi). Jak jsme si již řekli, tlak je **izotropní**: působí ve všech směrech stejně a vždy působí kolmo (normálově) na daný povrch. 

### Důkaz izotropie tlaku

#### 1. Výchozí předpoklad: Absence smykového napětí

Základní definice tekutiny říká, že tekutina v klidu nemůže vzdorovat (a tedy ani přenášet) smykovému napětí. Z toho přímo vyplývá, že pro **jakoukoliv** rovinu proloženou uvažovaným bodem v tekutině musí být smykové napětí nulové:
$$ \tau = 0 $$

#### 2. Vlastnosti Mohrovy kružnice

Mohrova kružnice je grafické znázornění tenzoru napětí v daném bodě. Každý bod na obvodu této kružnice reprezentuje stav napětí na jedné konkrétní rovině (pod určitým úhlem) proložené daným bodem materiálu.
*   **Horizontální osa:** Vynáší se na ni normálové napětí ($\sigma$).
*   **Vertikální osa:** Vynáší se na ni smykové napětí ($\tau$).

Střed kružnice ($S$) a její poloměr ($R$) jsou v 2D napjatosti dány vztahy:
$$ S = \frac{\sigma_x + \sigma_z}{2} $$
$$ R = \sqrt{\left(\frac{\sigma_x - \sigma_z}{2}\right)^2 + \tau_{xz}^2} $$

#### 3. Aplikace na tekutinu v klidu

Jelikož v tekutině v klidu neexistují žádná smyková napětí na žádné rovině, musí mít každý bod ležící na obvodu Mohrovy kružnice nulovou $y$-ovou souřadnici ($\tau = 0$).

Geometricky existuje pouze jediný způsob, jak může "kružnice" ležet celá na horizontální ose a nemít žádné body s nenulovou vertikální souřadnicí: **musí mít nulový poloměr ($R = 0$).**

Kružnice o nulovém poloměru degraduje na jediný bod na ose normálových napětí ($\sigma$).

#### 4. Analytický a fyzikální závěr
Pokud dosadíme náš předpoklad ($\tau_{xz} = 0$ pro uvažovaný element) do rovnice pro poloměr a víme, že i poloměr musí být nulový ($R = 0$, aby nevznikla normálová rovina s nenulovým smykem), dostaneme:

$$ 0 = \sqrt{\left(\frac{\sigma_x - \sigma_z}{2}\right)^2 + 0} $$

Tato rovnice má jediné řešení:
$$ \sigma_x - \sigma_z = 0 \implies \sigma_x = \sigma_z $$

(V trojrozměrném prostoru bychom logiku zopakovali pro všechny roviny a získali bychom $\sigma_x = \sigma_y = \sigma_z$).

V mechanice tekutin definujeme hydrostatický tlak $p$ jako tlakové normálové napětí vzaté s opačným znaménkem (protože tlak je komprese, zatímco v mechanice kontinua je za kladné $\sigma$ považován tah):
$$ p = -\sigma $$

Z naší rovnice tedy plyne:
$$ -p_x = -p_z \implies p_x = p_z $$

**Závěr:** Mohrova kružnice pro tekutinu v klidu je bod. To znamená, že nezávisle na tom, jak natočíme zkoumanou rovinu (jaký úhel na kružnici zvolíme), hodnota normálového napětí (a tedy i tlaku) zůstává naprosto stejná a nevzniká žádný smyk. **Tlak je tedy izotropní.**
Pojďme si dokázat, že tlak působí všemi směry stejně. Budeme uvažovat malý elementární objem tekutiny ve tvaru pravoúhlého trojúhelníku o výšce $dz$, délce $dx$ a jednotkové šířce $b$ (šířka směřuje "do papíru").

### Rovnice hydrostatiky 

Pojďme zjistit, jak prostorová změna tlaku způsobuje výslednou sílu. Představme si malý hranol tekutiny o rozměrech $dx, dy, dz$. Osu $z$ definujeme jako kladnou směrem nahoru.

Předpokládejme, že jediné síly působící na hranol jsou gravitace a gradient tlaku. Označme tlak na spodní stěně jako referenční tlak $p_{bot} = p_0$. Tlak na horní stěně ($p_{top}$) pak můžeme vyjádřit pomocí rozvoje Taylorovy řady:

$$ p_{top} = p_0 + \left(\frac{dp}{dz}\right)_0 (\delta z) + \frac{1}{2!} \left(\frac{d^2p}{dz^2}\right)_0 (\delta z)^2 + ... $$

Pokud limitně zmenšíme $\delta z$ k nule, můžeme členy vyšších řádů zanedbat a ponechat si jen první derivaci. Výsledná síla na tento element tekutiny způsobená rozdílem tlaků na horní a spodní ploše je:

$$ F_{press} = (p_{bot} - p_{top}) dx dy = \left(p_0 - \left(p_0 + \frac{dp}{dz} dz\right)\right) dx dy = -\frac{dp}{dz} dx dy dz  $$

Tuto sílu $F_{press}$ definujeme tak, že působí v kladném směru osy $z$ (znaménko minus znamená, že pokud tlak s výškou roste, tlaková síla působí dolů).
Kromě tlaku působí na element tekutiny síla gravitační (tíha elementu). Hmotnost elementu je hustota $\rho$ krát objem $V$. Tíha směřuje dolů (do záporného směru $z$):

$$ F_{weight} = -mg = -\rho V g = -\rho g (dx dy dz)$$

Pro statickou rovnováhu musí být součet tlakové a gravitační síly roven nule:
$$ F_{press} + F_{weight} = 0 \quad (2.9) $$
$$ -\frac{dp}{dz} (dx dy dz) - \rho g (dx dy dz) = 0  $$

Jelikož rovnice platí pro jakýkoli objem, můžeme elementární objem $(dx dy dz)$ vykrátit a získáme **základní rovnici hydrostatiky**:

$$ \frac{dp}{dz} = -\rho g  $$

*(Dovysvětlení: Znaménko minus je zde proto, že osu $z$ jsme definovali jako kladnou směrem nahoru. To znamená, že s rostoucí nadmořskou výškou $z$ se tlak snižuje. A naopak, čím hlouběji pod hladinu jdeme (z klesá), tím tlak roste. Pokud bychom osu $z$ namířili dolů, po směru gravitace, znaménko minus by zmizelo.)*

V obecném 3D prostoru můžeme rovnici zapsat pomocí operátoru gradientu ($\nabla$):
$$ \nabla p = \frac{dp}{dx}\hat{i} + \frac{dp}{dy}\hat{j} + \frac{dp}{dz}\hat{k} \quad (2.12) $$
$$ \nabla p = -\rho g \hat{k} \quad (2.13) $$

Gradient tlaku je vždy v rovnováze s gravitací, zrychlením, viskózními silami a dalšími vnějšími silami.

### Přetlak a podtlak (Gauge and vacuum pressure)

Tlak se běžně udává jedním ze dvou způsobů:
1.  **Absolutní tlak:** Celkový tlak měřený od absolutní nuly (dokonalého vakua).
2.  **Relativní tlak:** Tlak vztažený k okolnímu (atmosférickému) tlaku $p_a$.
    *   Pokud je $p > p_a$, jedná se o **přetlak (Gage pressure)**: $p_{gage} = p - p_a$
    *   Pokud je $p < p_a$, jedná se o **podtlak (Vacuum pressure)**: $p_{vacuum} = p_a - p$

Většina měřicích přístrojů (manometrů) měří relativní tlak vzhledem k atmosféře. Pro výpočet absolutního tlaku musíte k naměřenému přetlaku vždy přičíst aktuální atmosférický tlak.



### Hydrostatická síla na svislou stěnu (Příklad)

Jak jsme si ukázali, tlaková změna v kapalné tekutině je dána vztahem $dp = -\rho g dz$. Integrujme tuto rovnici podél osy $z$ (od hladiny, kde $z=0$, do hloubky $z$):

$$ \int_{p_a}^{p} dp = -\int_{0}^{z} \rho g dz \quad (2.17) $$
$$ p - p_a = -\rho g z \Rightarrow p = p_a - \rho g z \quad (2.19) $$

*(Dovysvětlení: Pozor na konvenci! Protože je osa z kladná nahoru a hladina je $z=0$, všechny body pod vodou mají zápornou souřadnici $z$. Proto součin $(-\rho g z)$ pro záporné $z$ vyjde kladně a celkový tlak pod hladinou roste.)*

Chceme-li zjistit celkovou horizontální sílu na svislou stěnu (např. přehradu) o šířce $w$, musíme si uvědomit, že tlak se s hloubkou mění. Elementární síla $dF_x$ působící na vodorovný proužek stěny o výšce $dz$ a šířce $w$ je:
$$ dF_x = p dA = (p_a - \rho g z) w dz \quad (2.21) $$

Stěna má z jedné strany vodu a z druhé strany vzduch. Musíme tedy uvažovat tlaky z obou stran.
*   Síla zleva (od vody): $dF_l = (p_a - \rho g z) w dz$
*   Síla zprava (od atmosféry): $dF_r = -p_a w dz$
*   Celková elementární síla: $dF_x = dF_l + dF_r = (p_a - \rho g z) w dz - p_a w dz = -\rho g z w dz$

Vliv atmosférického tlaku se odečte. Celkovou sílu $F$ na stěnu od hladiny ($z=0$) až na dno ($z=-H$) získáme integrací:

$$ F = \int_{0}^{-H} -\rho g z w dz = \left[ -\frac{1}{2}\rho g w z^2 \right]_{0}^{-H} = -\frac{1}{2}\rho g w (-H)^2 = \frac{1}{2} \rho g w H^2 \quad (2.25) $$

Tato celková síla však nepůsobí v těžišti stěny, ale níže, protože tlak u dna je mnohem větší než tlak u hladiny. Místo, kde síla působí, se nazývá **působiště tlaku (Center of pressure)** a spočítáme jej pomocí momentové rovnováhy.
Elementární moment síly k bodu $0$ na hladině je síla krát rameno $z$:
$$ dM_0 = z \cdot dF_x = z (-\rho g z w dz) \quad (2.29) $$

Integrací přes celou hloubku získáme celkový moment $M_0$:
$$ M_0 = \int_{0}^{-H} -\rho g w z^2 dz = \left[ -\frac{1}{3}\rho g w z^3 \right]_{0}^{-H} = \frac{1}{3} \rho g w H^3 $$ *(Poznámka: znaménko minus $M_0 = -\frac{1}{3}\rho g w H^3$, závisí na konvenci orientace momentu, absolutní velikost momentu je zásadní).*

Podle definice musí být celkový moment $M_0$ roven celkové síle $F$ vynásobené vzdáleností působiště $\bar{z}$:
$$ \bar{z} \cdot F = M_0 \Rightarrow \bar{z} = \frac{M_0}{F} $$
$$ \bar{z} = \frac{-1/3 \rho g w H^3}{1/2 \rho g w H^2} = -\frac{2}{3} H \quad (2.32) $$

**Výsledek:** Celková hydrostatická síla působí přesně ve **dvou třetinách hloubky** (měřeno od hladiny dolů).
