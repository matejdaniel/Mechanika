# Mechanika tekutin

**Mechanika tekutin** je podobor klasické mechaniky, který se zabývá studiem chování tekutin 
(kapalin a plynů) v klidu (hydrostatika a aerostatika) a v pohybu (hydrodynamika a aerodynamika), 
a silami, které vznikají při jejich interakci s pevnými tělesy nebo ohraničeními.

## Základní vlastnosti tekutin 

### Co definuje tekutinu?

**Tekutina** je látka, která se neustále deformuje (teče), je-li vystavena jakémukoli, i sebemenšímu, smykovému napětí. Na rozdíl od pevných látek nemají tekutiny vlastní tvar a vždy se přizpůsobují tvaru nádoby. Pojem „tekutina“ je souhrnný název zahrnující jak kapaliny, tak plyny.

#### Tekutina vs. Pevná látka

*   **Pevné látky v klidu:** Deformují se pod vlivem vnějších sil pouze do určité míry (odpovídají na napětí statickou deformací).
*   **Tekutiny v klidu:** Nemohou odolávat smykovému napětí. Aby tekutina mohla přenášet smykové napětí, musí být v pohybu (musí téct).

**Podobnosti:**
Pro obě skupenství (pevné i tekuté) využíváme tzv. hypotézu kontinua.
Na obě se vztahují základní fyzikální zákony mechaniky:
*   Zákon zachování hmotnosti
*   Newtonovy pohybové zákony (zákon zachování hybnosti)
*   První termodynamický zákon (zákon zachování energie)

Obě skupiny mají definované konstitutivní vztahy, které popisují vztah mezi napětím a rychlostí deformace (případně samotnou deformací).

#### Kapalina vs. Plyn

Ačkoliv obojí řadíme mezi tekutiny, vykazují zcela odlišné makroskopické chování:
*   **Kapalina:** Podléhá gravitaci, usazuje se na dně nádoby a vytváří tzv. volnou hladinu. Kapaliny jsou relativně obtížně stlačitelné (pro mnoho inženýrských výpočtů se uvažují jako nestlačitelné).
*   **Plyn:** Rychle a rovnoměrně vyplní celý prostor nádoby, ve které je uzavřen. Volnou hladinu netvoří. Plyny jsou vysoce stlačitelné a jejich objem se výrazně mění se změnou tlaku a teploty.

#### Ideální vs. Reálná tekutina

Abychom mohli matematicky popsat složité chování tekutin, zavádí mechanika rozdělení na ideální a reálné modely:

*   **Ideální kapalina:** Je to teoretický model tekutiny, která je **zcela nestlačitelná** a nemá **žádné vnitřní tření** (viskozita je nulová). Při jejím proudění nevzniká žádné smykové napětí, nedochází k disipaci energie na teplo a proudění je dokonale hladké.

*   **Reálná kapalina):** Vykazuje **vnitřní tření (viskozitu)**. Při pohybu reálné tekutiny na sebe jednotlivé vrstvy působí tečnými silami, což způsobuje odpor proti proudění a ztráty mechanické energie (mění se na teplo). Navíc je každá reálná tekutina (včetně kapalin) do určité míry stlačitelná.

### Hypotéza kontinua

Tekutina se skládá z obrovského množství molekul, které se vůči sobě neustále a volně pohybují. Tento chaotický mikroskopický pohyb je v makroskopických procesech (kterými se budeme v tomto kurzu zabývat) nepostřehnutelný. 

Rozložení fyzikálních vlastností v tekutině je tak plynulé, že k analýze látky můžeme bez problémů použít diferenciální počet. Z tohoto důvodu budeme tekutinu nadále považovat za spojité prostředí. Tento předpoklad nazýváme **hypotéza kontinua**.

Představme si makroskopický objem tekutiny $\Delta V$, který má hmotnost $\Delta m$. Průměrná hustota tohoto objemu je $\Delta m / \Delta V$. Předpoklad kontinua nám umožňuje provést limitní přechod, kdy se tento objem blíží k nule (čímž náš objem rozdělujeme na menší a menší části). Hustota tekutiny v jednom konkrétním bodě je pak definována jako:

$$ \rho \equiv \lim_{\Delta V \to 0} \frac{\Delta m}{\Delta V} $$


> **Poznámka k platnosti kontinua:**
> Hypotéza kontinua selhává ve chvíli, kdy se zkoumaný objem stane natolik malým, že je srovnatelný s volnou dráhou molekul (vzdálenost, kterou molekula urazí, než narazí do jiné). Pro vzduch na úrovni hladiny moře je tato vzdálenost řádově $10^{-7}$ metru. Pokud klesneme pod tuto úroveň, musíme přejít od mechaniky tekutin ke statistické mechanice.

V celém kurzu budeme často mluvit o **částici tekutiny**. Není tím myšlena jedna molekula, ale velmi malý element tekutiny – dostatečně malý na to, aby šlo z matematického hlediska o bod, ale stále dostatečně velký, aby obsahoval dostatek molekul pro zachování vlastností kontinua.
