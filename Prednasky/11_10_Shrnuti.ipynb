## Shrnutí: Zákony termodynamiky

Na závěr našeho úvodu do inženýrské termodynamiky je užitečné si přehledně shrnout základní pilíře, na kterých celý tento vědní obor stojí. Tyto fundamentální zákony (označované čísly 0 až 3) nebyly odvozeny z žádné nadřazené teorie; vycházejí čistě z lidské zkušenosti a experimentálního pozorování přírody. Slouží jako absolutní pravidla pro chování energie a hmoty.


### Nultý termodynamický zákon (Zákon o teplotě)

Ačkoliv byl formulován až jako poslední (ve 30. letech 20. století), tvoří logický základ pro všechny ostatní, a proto dostal číslo nula. Zavádí objektivní pojem **teploty** a ospravedlňuje používání teploměrů.

> **Definice:** Pokud jsou dva termodynamické systémy (A a B) v tepelné rovnováze se systémem třetím (C), pak jsou nutně v tepelné rovnováze i navzájem.

* **Inženýrský význam:** Znamená to, že existuje určitá stavová veličina – teplota. Pokud mají dvě tělesa stejnou teplotu, jsou v tepelné rovnováze a po jejich fyzickém spojení mezi nimi nebude proudit žádné teplo. Systém "C" v tomto případě funguje jako náš teploměr.

### První termodynamický zákon (Zákon zachování energie)

Tento zákon je účetní knihou vesmíru. Definuje vztah mezi teplem, prací a **vnitřní energií** systému. Říká, že energie se nedá vyrobit z ničeho ani beze stopy zničit.

> **Definice:** Celková energie izolovaného systému zůstává konstantní. Energii lze pouze přeměňovat z jedné formy na druhou nebo ji přenášet mezi systémem a okolím.

* **Matematický zápis pro uzavřený systém:** $\Delta E = Q - W$
* **Inženýrský význam:** Z jakéhokoliv stroje nemůžeme získat více energie, než do něj vložíme (neexistuje perpetuum mobile 1. druhu). Zákon nám dává rovnice energetické bilance pro výpočet výkonu kompresorů, dodaného tepla v kotlích a práce turbín.

### Druhý termodynamický zákon (Zákon o směru a entropii)

Zatímco první zákon povoluje jakoukoliv přeměnu energie, pokud sedí její celková bilance, druhý zákon zavádí do přírody "jednosměrku". Určuje směr přirozených procesů a definuje limitní účinnost přeměny tepla na práci. Je s ním spojena veličina zvaná **entropie**.

Druhý zákon má několik ekvivalentních formulací:

> **Clausiova formulace:** Není možné sestrojit cyklicky pracující zařízení, jehož jediným výsledkem by byl přenos tepla z chladnějšího tělesa na těleso teplejší. (Teplo samovolně teče jen z teplého na studené).
> **Kelvin-Planckova formulace:** Není možné sestrojit cyklicky pracující tepelný motor, který by odebíral teplo z jediné lázně a beze zbytku ho přeměnil na práci. (Každý motor musí část tepla odvést do chladiče, účinnost je vždy $< 100\,\%$).

* **Zákon rostoucí entropie:** V izolovaném systému entropie (míra neuspořádanosti a znehodnocení energie) nikdy neklesá. Při vratných dějích zůstává konstantní, při reálných (nevratných) dějích vždy roste.
* **Inženýrský význam:** Dává nám Carnotovu účinnost jakožto teoretický strop pro jakoukoliv elektrárnu či motor. Pomáhá nám identifikovat ztráty (tření, sdílení tepla) a definuje exergii (kvalitu energie). Zakazuje perpetuum mobile 2. druhu.

### Třetí termodynamický zákon (Zákon absolutní nuly)

Třetí zákon (Nernstův teorém) se zabývá chováním látek při extrémně nízkých teplotách a dává entropii absolutní referenční bod.

> **Definice:** Jak se teplota systému blíží k absolutní nule (0 K), entropie systému se blíží ke konstantní minimální hodnotě. U dokonalého krystalu čisté látky je tato entropie při absolutní nule přesně rovna nule.

* **Důsledek:** Absolutní nuly (0 K, tedy -273,15 °C) nelze dosáhnout žádným procesem s konečným počtem kroků.
* **Inženýrský význam:** Umožňuje chemikům a fyzikům vypočítat absolutní hodnoty entropie chemických látek, což je klíčové pro určování chemické rovnováhy a analýzu spalovacích nebo jiných chemických reakcí.
