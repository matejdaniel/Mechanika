## Základy statistické mechaniky

**Statistická mechanika** představuje most mezi tímto makroskopickým světem a mikroskopickým světem atomů a molekul. Místo abychom sledovali každou částici zvlášť, používáme teorii pravděpodobnosti a statistiky k určení průměrného chování celého systému.

Abychom tento přechod pochopili, musíme si definovat dva naprosto klíčové pojmy: mikrostav a makrostav.


### Makrostav a Mikrostav

> **Makrostav** je celkový, makroskopický popis systému definovaný pomocí pozorovatelných vlastností, jako jsou celkový počet částic ($N$), celkový objem ($V$), celková vnitřní energie ($U$), tlak ($p$) nebo teplota ($T$).

> **Mikrostav** je naproti tomu naprosto detailní, mikroskopický popis systému. Pro klasický plyn to znamená přesnou znalost polohy a hybnosti (rychlosti) každé jedné částice v daném okamžiku. V kvantové mechanice odpovídá mikrostav konkrétní vlnové funkci nebo souboru kvantových čísel celého systému.

Jeden určitý makrostav může být realizován obrovským množstvím různých mikrostavů. Pokud přehodíte dvě identické molekuly kyslíku ve vaší místnosti, získáte nový mikrostav, ale makrostav (teplota a tlak v místnosti) zůstane naprosto stejný.

Základní postulát statistické mechaniky zní: **Pro izolovaný systém v rovnováze jsou všechny jeho dostupné mikrostavy (odpovídající dané celkové energii) stejně pravděpodobné.**

### Příklad: Nádoba rozdělená na dvě části

Představme si jednoduchý modelový systém: Uzavřenou pevnou nádobu, která je pomyslnou přepážkou rozdělena na dvě stejně velké poloviny – levou (L) a pravou (R). V nádobě je celkem $N$ rozlišitelných částic.

* **Makrostav** je v tomto případě určen pouze tím, kolik částic je v levé polovině ($N_L$) a kolik v pravé polovině ($N_R$). Platí, že $N_L + N_R = N$.
* **Mikrostav** je dán přesným výčtem toho, *která konkrétní* částice se nachází v levé polovině a která v pravé.

Podívejme se na situaci pro $N = 4$ částice (označme je A, B, C, D):

1. **Makrostav $N_L = 4, N_R = 0$:**
Všechny částice jsou vlevo. Existuje pouze **1 mikrostav** (A, B, C, D jsou vlevo).
2. **Makrostav $N_L = 3, N_R = 1$:**
Tři vlevo, jedna vpravo. Vpravo může být A, B, C, nebo D. Tento makrostav lze realizovat **4 mikrostavy**.
3. **Makrostav $N_L = 2, N_R = 2$:**
Dvě vlevo, dvě vpravo. Možné dvojice vlevo jsou: AB, AC, AD, BC, BD, CD. Máme **6 mikrostavů**.

Počet mikrostavů $\Omega$ (čti omega), kterými lze realizovat daný makrostav, se nazývá **termodynamická pravděpodobnost** nebo statistická váha. Vypočítá se pomocí kombinatoriky (kombinačního čísla):

$$ \Omega = \frac{N!}{N_L! N_R!} $$

Z našeho příkladu (1, 4, 6, 4, 1) vidíme, že nejpravděpodobnější je ten makrostav, který má nejvyšší počet mikrostavů. V případě 4 částic je to stav rovnoměrného rozdělení (2 vlevo, 2 vpravo).

Pokud bychom zvýšili počet částic na $N = 100$ nebo na hodnotu blízkou Avogadrově konstantě ($N \approx 10^{23}$), distribuce počtu mikrostavů se stane extrémně ostrou špičkou. Drtivá většina všech myslitelných mikrostavů bude odpovídat makrostavu $N_L \approx N_R \approx N/2$.

Pokud systém začne v nerovnovážném stavu (všechny částice vlevo) a odstraníme přepážku, systém bude náhodně procházet mikrostavy. Protože mikrostavů odpovídajících rovnoměrnému rozložení je astronomicky více, systém přirozeně a samovolně přejde do makrostavu rovnoměrného rozdělení. **Tohle je mikroskopická podstata nevratnosti a druhého termodynamického zákona.**

### Odvození Boltzmannovy definice entropie

Nyní si ukažme, jak z počtu mikrostavů $\Omega$ získáme nejdůležitější stavovou veličinu statistické termodynamiky – entropii. Představme si dva izolované systémy $A$ a $B$.

* Systém $A$ má energii $E_A$ a počet mikrostavů $\Omega_A(E_A)$.
* Systém $B$ má energii $E_B$ a počet mikrostavů $\Omega_B(E_B)$.

Nyní tyto dva systémy uvedeme do tepelného kontaktu (mohou si vyměňovat energii, nikoliv částice nebo objem) a vytvoříme celkový izolovaný systém $A+B$. Celková energie je konstantní:

$$ E_{tot} = E_A + E_B = \text{konst.} $$

Protože systémy jsou nezávislé, celkový počet mikrostavů spojeného systému $\Omega_{tot}$ je roven součinu počtu mikrostavů jednotlivých systémů:

$$ \Omega_{tot}(E_A, E_B) = \Omega_A(E_A) \cdot \Omega_B(E_B) $$

Podle základního postulátu se systém vyvine do nejpravděpodobnějšího makrostavu, tedy do rovnováhy. Rovnovážný stav je ten, který maximalizuje počet dostupných mikrostavů $\Omega_{tot}$. Místo samotné funkce $\Omega$ je matematicky výhodnější maximalizovat její přirozený logaritmus, $\ln \Omega_{tot}$, protože z maximalizace logaritmu vyplývá totéž (logaritmus je rostoucí funkce) a navíc převádí součin na součet:

$$ \ln \Omega_{tot} = \ln \Omega_A + \ln \Omega_B $$

V maximu musí platit, že derivace podle předávané energie je nulová:

$$ d(\ln \Omega_{tot}) = \left( \frac{\partial \ln \Omega_A}{\partial E_A} \right) dE_A + \left( \frac{\partial \ln \Omega_B}{\partial E_B} \right) dE_B = 0 $$

Zároveň ze zákona zachování energie víme, že co systém $A$ přijme, to systém $B$ odevzdá, takže $dE_A = -dE_B$. Dosazením získáme:

$$ \left( \frac{\partial \ln \Omega_A}{\partial E_A} \right) dE_A - \left( \frac{\partial \ln \Omega_B}{\partial E_B} \right) dE_A = 0 $$

$$ \frac{\partial \ln \Omega_A}{\partial E_A} = \frac{\partial \ln \Omega_B}{\partial E_B} $$

Tato rovnice nám říká úžasnou věc: v tepelné rovnováze se určitá veličina závislá na změně mikrostavů vůči energii musí pro oba systémy rovnat. Z makroskopické termodynamiky (nultý zákon) víme, že veličina, která se v tepelné rovnováze rovná, je **teplota** $T$. Veličinu $\frac{\partial \ln \Omega}{\partial E}$ tedy musíme definovat tak, aby byla úměrná $1/T$. Historicky se zavedla Boltzmannova konstanta $k_B$:

$$ \frac{\partial \ln \Omega}{\partial E} = \frac{1}{k_B T} $$

Z klasické termodynamiky také známe vztah pro entropii $S$ při stálém objemu a počtu částic: $dS = \frac{dE}{T}$, tedy $\frac{\partial S}{\partial E} = \frac{1}{T}$. Porovnáním obou vztahů získáme slavnou Boltzmannovu rovnici pro entropii:

$$ S = k_B \ln \Omega $$

Entropie $S$ je tedy logaritmickou mírou počtu mikrostavů $\Omega$, kterými lze realizovat daný makrostav. Zvyšování entropie (dle druhého termodynamického zákona) není nic jiného než přechod systému do makrostavu, který je realizován mnohem větším počtem mikrostavů (je statisticky pravděpodobnější).

### Nultý termodynamický zákon

V předcházejícím textu jsme předpokládali rovnost teploty u systémů v kontaktu. Tato rovnost, se označuje také jako **nultý termodynamický zákon**.

> **Nultý termodynamický zákon zní:**
> Pokud jsou dva termodynamické systémy (A a B) v tepelné rovnováze se systémem třetím (C), pak jsou nutně v tepelné rovnováze i navzájem (systém A je v rovnováze se systémem B).

Matematicky bychom to mohli zapsat jako vlastnost tranzitivity:
Pokud platí $A \sim C$ a současně $B \sim C$, pak musí platit i $A \sim B$.

#### Proč se nazývá "nultý"?

Historicky byly první a druhý termodynamický zákon formulovány a všeobecně přijaty již v 19. století. Teprve na počátku 20. století (často se uvádí rok 1931 a fyzik Ralph H. Fowler) si vědci uvědomili, že pro logickou úplnost termodynamiky je potřeba zavést postulát, který definuje teplotu a ospravedlňuje použití teploměrů. Protože tento zákon tvoří absolutní logický základ pro zbylé zákony, ale ty už měly svá zavedená čísla (1, 2, 3), byl nazván nultým zákonem.

#### Zásadní inženýrský a fyzikální význam

Na první pohled se může nultý zákon zdát jako triviální selský rozum. Jeho skutečný význam ale spočívá v tom, že nám dává **objektivní způsob, jak měřit teplotu**.

Představte si následující situaci z praxe:

1. Máte motor (systém A) a chladicí kapalinu (systém B). Chcete vědět, zda jsou ve stejné tepelné kondici (zda si nebudou předávat teplo).
2. Nemůžete je jednoduše "přiložit k sobě" a čekat, co se stane. Místo toho použijete **teploměr (systém C)**.
3. Teploměr (C) nejprve přiložíte k motoru (A). Počkáte, až se vlastnosti teploměru přestanou měnit – rtuť se zastaví na určité rysce. Systém A a C jsou nyní v tepelné rovnováze.
4. Poté teploměr (C) vložíte do chladicí kapaliny (B). Pokud rtuť ukazuje přesně stejnou hodnotu, znamená to, že kapalina (B) je s teploměrem (C) ve stejné tepelné rovnováze jako motor (A).
5. Z nultého zákona tedy jasně vyplývá: motor (A) a kapalina (B) jsou v tepelné rovnováze navzájem. Pokud byste je dostali do přímého kontaktu, žádné teplo by mezi nimi neteklo.

Nultý zákon nám tak říká, že tepelná rovnováha je ekvivalence a že existuje jakási univerzální vlastnost systému, která určuje, zda je systém s jiným v rovnováze. Tuto vlastnost nazýváme **teplota**. Nultý zákon je tedy samotnou definicí teploměru a garantuje nám, že měření teploty má vůbec fyzikální smysl.


### Důkaz maxima entropie v rovnovážném stavu

V předchozím odvození jsme zjistili, že v rovnovážném stavu platí rovnost derivací, tedy $d(\ln \Omega_{tot}) = 0$. Z matematické analýzy víme, že nulová první derivace značí extrém (minimum, maximum nebo inflexní bod). Jak ale s jistotou dokážeme, že se jedná právě o **maximum** celkové entropie, a nikoliv o minimum?

Tento důkaz můžeme provést dvěma způsoby: analýzou samovolného přenosu tepla nebo matematicky pomocí druhé derivace. Obě cesty nám navíc propojí statistickou mechaniku s druhým termodynamickým zákonem.

#### 1. Fyzikální důkaz: Směr přenosu tepla a růst entropie

Představme si, že naše dva systémy $A$ a $B$ ještě nejsou v rovnováze. Uveďme je do tepelného kontaktu. Předpokládejme, že systém $B$ je teplejší než systém $A$ (tedy $T_B > T_A$). Zkušenost i termodynamika nám říkají, že teplo (energie) poteče samovolně z teplejšího tělesa na chladnější.

Zkoumejme nyní celkovou změnu entropie spojeného systému $dS_{tot}$ při tomto malém přenosu energie. Celková změna entropie je součtem změn v obou systémech:

$$ dS_{tot} = dS_A + dS_B $$

Zároveň víme, že entropie je funkcí vnitřní energie, můžeme tedy zapsat její úplný diferenciál (při konstantním objemu a počtu částic):

$$ dS_{tot} = \left( \frac{\partial S_A}{\partial E_A} \right) dE_A + \left( \frac{\partial S_B}{\partial E_B} \right) dE_B $$

Protože celkový systém $A+B$ je izolovaný, energie, kterou systém $A$ získá, musí být rovna energii, kterou systém $B$ ztratí. Platí tedy zákon zachování energie v diferenciálním tvaru: $dE_B = -dE_A$. Dále využijeme definici teploty, kterou jsme si odvodili: $\frac{\partial S}{\partial E} = \frac{1}{T}$.

Dosazením těchto dvou vztahů dostáváme:

$$ dS_{tot} = \frac{1}{T_A} dE_A - \frac{1}{T_B} dE_A = \left( \frac{1}{T_A} - \frac{1}{T_B} \right) dE_A $$

Nyní aplikujeme náš počáteční předpoklad. Teplo teče ze systému $B$ do systému $A$, takže energie systému $A$ roste ($dE_A > 0$). Současně víme, že $T_B > T_A$. Z toho matematicky vyplývá, že $\frac{1}{T_A} > \frac{1}{T_B}$, a tedy závorka $\left( \frac{1}{T_A} - \frac{1}{T_B} \right)$ musí být **kladné číslo**.

Vynásobením dvou kladných hodnot dostáváme jasný výsledek:

$$ dS_{tot} > 0 $$

**Závěr:** Jakýkoliv samovolný proces, při kterém si systémy vyměňují energii a blíží se k rovnováze, vede k růstu celkové entropie. Jakmile se teploty vyrovnají ($T_A = T_B$), člen v závorce bude nulový a růst entropie se zastaví ($dS_{tot} = 0$). Protože entropie neustále roste, dokud se systémy nevyrovnají, z logiky věci vyplývá, že rovnovážný stav ($T_A = T_B$) je stavem s absolutně **maximální celkovou entropií**.

> Tento výsledek není nic jiného než **Druhý termodynamický zákon** odvozený ze statistických principů: V izolovaném systému entropie s časem neklesá, naopak roste, dokud nedosáhne maxima v termodynamické rovnováze.

#### 2. Matematický důkaz: Podmínka druhé derivace

Pokud chceme být matematicky naprosto precizní, musíme ukázat, že druhá derivace celkové entropie podle přenášené energie je v bodě extrému záporná. Podívejme se na druhou derivaci entropie izolovaného podsystému podle jeho energie:

$$ \frac{\partial^2 S}{\partial E^2} = \frac{\partial}{\partial E} \left( \frac{1}{T} \right) $$

Použitím pravidla pro derivování složených funkcí získáme:

$$ \frac{\partial}{\partial E} \left( \frac{1}{T} \right) = -\frac{1}{T^2} \frac{\partial T}{\partial E} $$

Závislost energie na teplotě udává fyzikální veličina zvaná **tepelná kapacita** (při stálém objemu, $C_v$). Její definice je $C_v = \frac{\partial E}{\partial T}$. Člen $\frac{\partial T}{\partial E}$ je tedy roven převrácené hodnotě tepelné kapacity, $\frac{1}{C_v}$.

Rovnice přejde do tvaru:

$$ \frac{\partial^2 S}{\partial E^2} = -\frac{1}{T^2 C_v} $$

Aby byl systém makroskopicky stabilní, musí platit, že když mu dodáme energii (teplo), jeho teplota stoupne. To znamená, že tepelná kapacita $C_v$ musí být vždy striktně kladná ($C_v > 0$). Teplota $T$ umocněná na druhou je také vždy kladná.

Z toho vyplývá:

$$ \frac{\partial^2 S}{\partial E^2} < 0 $$

Druhá derivace entropie podle energie je vždy záporná. Zkoumaný extrém (první derivace rovna nule) pro náš spojený systém je tedy s naprostou matematickou jistotou **ostré maximum**. Systém vždy přirozeně diverguje do makrostavu, který odpovídá maximálnímu počtu mikrostavů $\Omega$, tedy maximální entropii.

### Avogadrova konstanta a pojem mol

Jak jsme viděli v předchozím textu, pro pochopení makroskopických termodynamických vlastností (jako je tlak nebo teplota) potřebujeme pracovat s obrovským množstvím částic. Abychom s těmito astronomickými čísly mohli v inženýrské a chemické praxi efektivně počítat, zavádíme veličinu zvanou **látkové množství** (jednotka mol) a s ní spojenou **Avogadrovu konstantu**.

> **Avogadrova konstanta ($N_A$)** udává přesný počet elementárních entit (atomů, molekul, iontů nebo jiných částic), které se nacházejí v jednom molu dané látky.

Od redefinice základních jednotek soustavy SI (platné od roku 2019) je Avogadrova konstanta definována jako přesné číslo:

$$ N_A = 6,022\,140\,76 \times 10^{23} \text{ mol}^{-1} $$

Jeden mol jakékoliv látky tedy obsahuje přesně $6,022\,140\,76 \times 10^{23}$ částic. Hmotnost jednoho molu látky (molární hmotnost) v gramech je přitom číselně zhruba rovna její relativní atomové nebo molekulové hmotnosti. Například jeden mol vody ($H_2O$) má hmotnost zhruba 18 gramů, což odpovídá objemu jedné polévkové lžíce.


#### Jak obrovské je to číslo? Příklady pro lepší představu

Lidský mozek není evolučně uzpůsoben k tomu, aby si dokázal představit číslo v řádu $10^{23}$. Jde o hodnotu tak extrémní, že ji nejlépe pochopíme pomocí přirovnání:

**1. Písek pokrývající planetu**
Představte si jedno zrnko písku jako malou krychličku o objemu $1 \text{ mm}^3$. Pokud byste vzali jeden mol těchto zrnek písku (tedy přesně $N_A$ zrníček) a rovnoměrně byste je rozsypali po úplně celém povrchu planety Země (včetně všech oceánů, pouští i hor), vytvořily by souvislou vrstvu písku o tloušťce přes **1,2 metru**.

**2. Utrácení peněz od Velkého třesku**
Pokud byste byli multimiliardáři a na vašem účtu by leželo přesně $6,022 \times 10^{23}$ českých korun. Kdybyste začali utrácet **1 milion korun každou sekundu**, každý den, bez jediné přestávky od okamžiku Velkého třesku (tedy po dobu zhruba 13,8 miliardy let), do dnešního dne byste neutratili ani celou Avogadrovu konstantu (utratili byste "jen" zhruba $4,3 \times 10^{23}$ korun).

**3. Věž z papíru**
Vezměte jeden list běžného kancelářského papíru (tloušťka zhruba 0,1 mm). Pokud byste naskládali na sebe $N_A$ takových listů, jak vysoká by byla výsledná věž? Věž by měřila zhruba $6 \times 10^{16} \text{ kilometrů}$. Pro představu, světlo putuje ze Slunce na Zemi 8 minut. K vrcholu této papírové věže by světlo letělo zhruba **6 300 let**. Tato hromada by bez problému přesáhla hranice naší Sluneční soustavy a dosáhla by hluboko do naší galaxie.

**4. Lžíce vody v oceánech**
Tento příklad krásně ilustruje mikroskopické měřítko atomů. Jak jsme zmínili, jedna polévková lžíce vody (18 ml) obsahuje zhruba $N_A$ molekul vody. Kdybyste tyto molekuly speciálně "obarvili", lžíci vylili do oceánu a počkali tisíce let, dokud se oceány na Zemi dokonale nepromíchají, co by se stalo? Ať už byste pak kdekoliv na světě nabrali z oceánu novou lžíci vody, našli byste v ní v průměru **desítky vašich původních, obarvených molekul**.
