## Kinetická energie, potenciální energie a termodynamická práce


### Práce a kinetická energie

Představme si těleso o hmotnosti $m$ (náš uzavřený systém), které se pohybuje po obecné křivce. Rychlost těžiště tohoto tělesa označíme $v$. Na těleso působí výsledná síla $\vec{F}$, která se může měnit jak v čase, tak v závislosti na poloze. Tuto sílu můžeme rozložit na tečnou složku $\vec{F_s}$ (působící ve směru dráhy) a normálovou složku $F_n$ (působící kolmo na dráhu). Tečná složka $\vec{F_s}$ mění velikost rychlosti, zatímco normálová složka $F_n$ mění její směr.

Okamžitou polohu tělesa na dráze označíme $s$. Podle druhého Newtonova pohybového zákona je velikost tečné síly $F_s$ úměrná zrychlení:

$$ F_s = m \frac{dv}{dt} $$

Pomocí pravidla pro derivace můžeme tento vztah upravit, přičemž využijeme faktu, že rychlost $V = \frac{ds}{dt}$:

$$ F_s = m \frac{dv}{ds} \frac{ds}{dt} = m v \frac{dv}{ds} $$

Tuto rovnici upravíme a budeme integrovat od počáteční polohy $s_1$ (kde je rychlost $v_1$) do konečné polohy $s_2$ (kde je rychlost $v_2$):

$$ \int_{s_1}^{s_2} F_s \, ds = \int_{v_1}^{v_2} m V \, dV $$

Integrál na pravé straně snadno vyhodnotíme:

$$ \int_{v_1}^{v_2} m V \, dV = \left[ \frac{1}{2} m v^2 \right]_{v_1}^{v_2} = \frac{1}{2} m v_2^2 - \frac{1}{2} m v_1^2 $$

Veličina $\frac{1}{2} m V^2$ se nazývá **kinetická energie** ($KE$). Jde o skalární veličinu. Změnu kinetické energie systému označíme jako $\Delta E_k$:

$$ \Delta E_k = E_{k2} - E_{k1} = \frac{1}{2} m (v_2^2 - v_1^2) $$

Levá strana naší původní integrální rovnice, tedy integrál síly po dráze, představuje mechanickou **práci** $W$, kterou síla $F_s$ vykonala na tělese. Celkově tedy dostáváme:

$$ W = \int_{s_1}^{s_2} F_s \, ds = \frac{1}{2} m (v_2^2 - v_1^2) = \Delta E_k $$

Tento vztah říká, že práce vykonaná výslednou silou na tělese se rovná změně jeho kinetické energie. Kinetická energie je *extenzivní vlastností* systému – závisí pouze na hmotnosti a okamžité rychlosti tělesa, bez ohledu na to, jak této rychlosti těleso dosáhlo.

### Potenciální energie

Nyní tento koncept rozšíříme. Představme si těleso o hmotnosti $m$, které se pohybuje vertikálně z výšky $z_1$ do výšky $z_2$ vzhledem k povrchu Země. Na systém působí dvě síly: gravitační síla směřující dolů o velikosti $mg$ a výslednice všech ostatních sil, kterou označíme $R$.

Podle předchozího odvození se celková práce vykonaná na tělese musí rovnat změně jeho kinetické energie. Celková práce je algebraickým součtem práce síly $R$ a práce gravitační síly (ta má záporné znaménko, protože působí proti směru rostoucí výšky $z$):

$$ \int_{z_1}^{z_2} R \, dz - \int_{z_1}^{z_2} mg \, dz = \frac{1}{2} m (v_2^2 - v_1^2) $$

Druhý integrál, představující práci gravitační síly, můžeme při předpokladu konstantního tíhového zrychlení $g$ snadno vyřešit jako $mg(z_2 - z_1)$. Rovnici pak můžeme přepsat do tvaru:

$$ \int_{z_1}^{z_2} R \, dz = \frac{1}{2} m (v_2^2 - v_1^2) + mg (z_2 - z_1) $$

Veličina $mgz$ se nazývá **gravitační potenciální energie** ($E_p$). Změna potenciální energie je tedy:

$$ \Delta E_p = E_{p2} - E_{p1} = mg (z_2 - z_1) $$

Podobně jako kinetická energie je i potenciální energie extenzivní vlastností systému (i když přesněji řečeno jde o vlastnost systému tvořeného tělesem a Zemí dohromady).


### Zobecnění pojmu práce v termodynamice

V mechanice jsme práci definovali makroskopicky jako působení síly po dráze:

$$ W = \int_{s_1}^{s_2} \mathbf{F} \cdot d\mathbf{s} $$

Termodynamika se však zabývá i jevy, které se do tohoto úzce mechanického rámce nevejdou. Proto musíme přijmout širší definici.

> **Termodynamická definice práce:**
> Interakce mezi systémem a okolím je klasifikována jako práce, pokud platí: *Jediným účinkem na vše, co je vně systému (v okolí může být zvednutí závaží.*

Zvednutí závaží je v podstatě síla působící po dráze. Termodynamická definice ale neříká, že se závaží muselo *skutečně* zvednout, stačí, že by to mohl být *jediný teoretický účinek* na okolí.

**Příklad z praxe:**
Představme si uzavřený systém, který tvoří pouze elektrická baterie. Baterie napájí vnější elektrický obvod s odporem (např. žárovku). Na hranici systému (svorkách baterie) nevidíme žádnou mechanickou sílu ani pohyb. Jde tedy o práci?
Ano. Pokud bychom totiž žárovku nahradili ideálním elektromotorem spojeným s navijákem, elektromotor by mohl zvedat závaží. Jediným teoretickým účinkem na okolí baterie by tedy mohlo být zvednutí závaží. Elektrická energie dodávaná baterií do okolí je proto z termodynamického hlediska **prací**.

### Znaménková konvence a značení

V inženýrské termodynamice se často zabýváme stroji, jejichž primárním účelem je práci *produkovat* (např. turbíny, spalovací motory). Proto zavádíme následující znaménkovou konvenci, které se budeme držet v celém textu:

* **$W > 0$:** Práci koná systém (systém odevzdává energii do okolí).
* **$W < 0$:** Práce je konána na systému (okolí dodává energii do systému).

### Práce není vlastnost (Neúplný diferenciál)

Hodnota vykonané práce $W$ zásadně závisí na tom, **jak** proces probíhal – závisí na detailech interakce mezi systémem a okolím, nejen na počátečním a konečném stavu. Z toho vyplývá, že **práce není vlastnost systému**.

Z matematického hlediska to znamená, že diferenciál práce označujeme speciálním symbolem $\delta W$ (čte se jako *neúplný diferenciál*), abychom ho odlišili od diferenciálů vlastností (např. $dV$ nebo $dp$), které jsou *úplné*. Nikdy nemůžeme psát $W_2 - W_1$, protože něco jako "práce ve stavu 1" neexistuje. Píšeme pouze $W_{12}$ nebo jednoduše $W$, což značí celkovou práci během procesu.

### Výkon

Rychlost, jakou je energie přenášena ve formě práce, se nazývá **výkon** (značíme $\dot{W}$, tečka nad písmenem značí časovou derivaci). Pro mechanickou sílu platí:

$$ \dot{W} = \vec{F} \cdot \vec{v} $$

Jednotkou výkonu v soustavě SI je watt (W) nebo kilowatt (kW). Celkovou práci za určitý čas získáme integrací výkonu: $W = \int \dot{W} \, dt$.


### Objemová práce při expanzi nebo kompresi

Jedním z nejdůležitějších způsobů, jak může systém konat práci, je změna objemu – expanze nebo komprese plynu (případně kapaliny).

Uvažujme plyn uzavřený ve válci s pohyblivým pístem o ploše průřezu $A$. Tlak plynu na čelo pístu označíme $p$. Síla, kterou plyn působí na píst, je $F = pA$. Pokud se píst posune o nekonečně malou vzdálenost $dx$, plyn vykoná elementární práci:

$$ \delta W = F \, dx = (pA) \, dx $$

Protože součin plochy a posunutí ($A \, dx$) odpovídá elementární změně objemu systému $dV$, získáváme klíčový vztah pro objemovou práci:

$$ \delta W = p \, dV $$

Pro proces z počátečního objemu $V_1$ do konečného objemu $V_2$ získáme celkovou práci integrací:

$$ W = \int_{V_1}^{V_2} p \, dV $$

*Při expanzi se objem zvětšuje ($dV > 0$) a práce konaná systémem je kladná. Při kompresi se objem zmenšuje ($dV < 0$) a práce je záporná.*


#### p-V diagram a polytropický děj

Abychom mohli integrál vyřešit, musíme znát funkci $p = f(V)$ – tedy to, jak se tlak mění s objemem během procesu. U skutečných rychlých procesů (např. ve válci motoru) není tlak v celém objemu stejný. Pokud ale uvažujeme **kvazistatický proces** (děj probíhající tak pomalu, že systém je v každém okamžiku v rovnováze), tlak je homogenní.

Závislost tlaku na objemu u kvazistatického procesu můžeme vynést do **p-V diagramu** (pracovního diagramu). Integrál $\int p \, dV$ pak geometricky představuje **plochu pod křivkou děje** v tomto diagramu. Z toho je zřejmé, že různé cesty mezi stejným počátečním a konečným stavem povedou k různě velkým plochám, a tedy i k různě velké práci.

Častým matematickým modelem pro reálné děje je tzv. **polytropický děj**, definovaný vztahem:

$$ pV^n = \text{konst.} $$

kde $n$ je polytropický exponent. Změnou $n$ můžeme popsat různé typy procesů.

### Zobecněná termodynamická práce

Jak jsme si uvedli v předchozí části, termodynamická práce se neomezuje pouze na expanzi a kompresi plynu v pístu ($W = \int p \, dV$). Tuto myšlenku můžeme zobecnit pro jakýkoliv fyzikální systém.

Každou formu termodynamické práce lze matematicky vyjádřit jako integrál součinu dvou sdružených veličin:


$$ W = \int_{X_1}^{X_2} Y \, dX $$

Kde:

* **$Y$ je zobecněná síla (intenzivní veličina):** Její hodnota nezávisí na velikosti systému. Udává "hnací sílu" nebo "potenciál" pro daný proces. V případě objemové práce to byl tlak $p$.
* **$dX$ je zobecněné posunutí (diferenciál extenzivní veličiny):** Její hodnota je přímo úměrná velikosti systému. Představuje "reakci" systému nebo "změnu rozsahu". V případě objemové práce to byla změna objemu $dV$.

Podívejme se nyní na konkrétní inženýrské a fyzikální příklady jiných forem práce a ukažme si u každého z nich rozdělení na intenzivní a extenzivní složku. *(Poznámka ke znaménkům: V následujících vztazích uvažujeme práci vykonanou okolím **na systému**, což z inženýrské konvence často označujeme záporným znaménkem, případně ji definujeme pro konkrétní úlohu kladně, pokud nás zajímá energie dodaná do systému).*


#### Práce způsobená mechanickým normálovým napětím (Pružná deformace)

Představme si pevnou tyč nebo drát o průřezu $A$ a původní délce $L_0$, který natahujeme mechanickou silou. V mechanice kontinua nevnímáme jen vnější sílu, ale vnitřní **normálové napětí** $\sigma = F/A$. Při natažení o malou délku $dL$ je vykonána objemová deformační práce. Protože součin průřezu a změny délky ($A \cdot dL$) představuje změnu deformovaného objemu $dV$ (případně můžeme pracovat s poměrným prodloužením), práce se dá vyjádřit pomocí napětí.

Integrál pro práci vykonanou *na systému* (natahování drátu) je:


$$ W = \int \sigma \cdot A \, dL = \int \sigma \, dV $$

* **Intenzivní veličina ($Y$):** Normálové napětí $\sigma$ [Pa]. Nezávisí na tom, jak je drát dlouhý.
* **Extenzivní veličina ($X$):** Deformovaný objem $V$ (nebo prostě délka drátu $L$ při uvažování konstantního průřezu) [m³] resp. [m].

#### Práce způsobená smykovým napětím

Tento případ nastává, když na pevnou látku působí tečné síly, které způsobují změnu tvaru (zkosení) bez nutnosti změny celkového objemu. Typickým příkladem je torze (kroucení) hřídele. Síla působící tečně k ploše vyvolává **smykové napětí** $\tau$. Deformace se pak měří pomocí smykového úhlu $\gamma$. Elementární práce vykonaná smykovým napětím na elementu objemu $V$ je úměrná velikosti zkosení $d\gamma$.

Integrál pro celkovou deformační práci je:


$$ W = \int \tau \cdot V \, d\gamma $$

* **Intenzivní veličina ($Y$):** Smykové napětí $\tau$ [Pa].
* **Extenzivní veličina ($X$):** Celková smyková výchylka (součin objemu a smykového úhlu $V\gamma$) [m³].

#### Práce povrchového napětí

Kapaliny se snaží minimalizovat svůj povrch v důsledku kohezních sil mezi molekulami. Chceme-li povrch kapaliny zvětšit (například při vyfukování mýdlové bubliny), musíme konat práci proti síle **povrchového napětí**. Pokud zvětšíme plochu povrchu kapaliny o element $dA$, vykonáme práci.

Integrál pro práci nutnou ke zvětšení povrchu (práce vykonaná na systému) je:


$$ W = \int \gamma \, dA $$


*(Někdy se povrchové napětí značí také symbolem $\sigma$).*

* **Intenzivní veličina ($Y$):** Povrchové napětí $\gamma$ [N/m] nebo [J/m²]. Záleží pouze na druhu kapaliny a teplotě, nezávisí na velikosti kapky.
* **Extenzivní veličina ($X$):** Celková plocha povrchu $A$ [m²].

#### Práce (energie) v elektrickém poli

Vložíme-li dielektrický materiál (izolant) do vnějšího elektrického pole, dojde k jeho polarizaci – molekuly se zorientují a vytvoří makroskopický elektrický dipól. Práce nutná ke změně polarizace systému v elektrickém poli souvisí s překonáváním tohoto pole.

Integrál pro elektrickou polarizační práci je:


$$ W = \int \mathbf{E} \cdot d\mathbf{p} $$


*(Kde tučně značené veličiny představují vektory a tečka je jejich skalární součin).*

* **Intenzivní veličina ($Y$):** Intenzita elektrického pole $\mathbf{E}$ [V/m]. (Případně elektrický potenciál $\varphi$ [V], pokud bychom uvažovali nabíjení kondenzátoru, kde $W = \int \varphi \, dq$).
* **Extenzivní veličina ($X$):** Celkový elektrický dipólový moment systému $\mathbf{p}$ [C·m]. (V případě kondenzátoru je extenzivní veličinou celkový elektrický náboj $q$ [C]).

#### Práce (energie) v magnetickém poli

Analogicky k elektrickému poli, pokud vložíme magnetický materiál (např. kus železa) do vnějšího magnetického pole, materiál se zmagnetizuje. Zvýšení magnetizace materiálu vyžaduje práci, která se následně uloží jako energie magnetického pole v daném objemu materiálu.

Integrál pro magnetickou polarizační práci je:


$$ W = \int \mathbf{H} \cdot d\mathbf{m} $$

* **Intenzivní veličina ($Y$):** Intenzita magnetického pole $\mathbf{H}$ [A/m]. (Tato veličina popisuje pole generované vnějšími proudy nezávisle na vzorku).
* **Extenzivní veličina ($X$):** Celkový magnetický moment materiálu $\mathbf{m}$ [A·m²]. (Závisí na tom, jak velký kus materiálu magnetizujeme).

#### Zobecněná síla a dráha

Pokud v systému probíhá současně více typů interakcí (např. systém je současně natahován a polarizován v elektrickém poli), celkovou elementární práci vyjádříme jako součet těchto příspěvků:

$$ \delta W = p \, dV - \sigma \, d(A_x) - \tau \, d\gamma - \gamma \, dA - \mathbf{E} \cdot d\mathbf{p} - \mathbf{H} \cdot d\mathbf{m} - \dots $$

Tento zápis je nesmírně silný, protože nám dává univerzální klíč: každá forma práce je vždy součinem určité **intenzivní veličiny** (která působí jako hnací síla) a diferenciálu s ní spojené **extenzivní veličiny** (která reprezentuje odezvu, změnu rozsahu nebo výchylku).

Proto se těmto dvojicím často říká **zobecněná síla** (např. tlak, elektrická intenzita, povrchové napětí) a **zobecněná dráha/posunutí** (např. objem, polarizace, plocha povrchu). Tento koncept překračuje hranice čisté mechaniky a spojuje celou fyziku pod jednu matematickou střechu.

*Poznámka:* Uvědomte si, že tato forma rovnice platí pouze pro *kvazistatické děje*. Existují typy práce (např. míchání kapaliny lopatkami nebo průchod elektrického proudu ohmickým odporem), které nelze jednoduše zapsat jako $Y \, dX$, protože při nich systém nutně prochází nerovnovážnými stavy. Zde se dostáváme k nevratnosti a druhému termodynamickému zákonu, což bude předmětem dalších kapitol.


##  Rozšíření pojmu energie (Vnitřní energie)

Dosud jsme z mechaniky znali dva druhy energie vázané na makroskopický systém jako celek:

1. **Kinetická energie ($E_k$)**: Vázaná na rychlost pohybu celého systému vůči referenční soustavě.
2. **Potenciální energie ($E_p$)**: Vázaná na polohu celého systému ve vnějším gravitačním poli.

Co se ale stane, když stlačíme pružinu a zaaretujeme ji? Nebo když nabijeme baterii? Nebo když izolovanou nádobu s plynem prudce mícháme lopatkami? V žádném z těchto případů se nemění rychlost těžiště systému ani jeho nadmořská výška. Kinetická i potenciální energie (ve smyslu vztahů z podkapitoly) jsou nulové. Přesto do systémů očividně dodáváme práci, která se tam musí nějak uložit. Z mechaniky víme, že energie se neztrácí.

Zde termodynamika rozšiřuje mechanický pohled a zavádí **vnitřní energii** ($U$).

> V inženýrské termodynamice se změna celkové energie systému ($\Delta E$) skládá ze tří částí: makroskopické kinetické energie ($\Delta E_k$), makroskopické potenciální energie ($\Delta E_p$) a **všech ostatních změn energie, které se sloučí do tzv. vnitřní energie ($\Delta U$)**.

$$ \Delta E = \Delta E_k + \Delta E_p + \Delta U $$

$$ E_2 - E_1 = (E_{k2} - E_{k1}) + (E_{p2} - E_{p1}) + (U_2 - U_1)$$

Stejně jako celková energie, i vnitřní energie $U$ je extenzivní vlastností. Často pracujeme s *měrnou vnitřní energií* $u$ (vztaženou na jednotku hmotnosti, např. kJ/kg) nebo s molární vnitřní energií (vztaženou na jednotku látkového množství).

### Mikroskopický původ vnitřní energie

Abychom si vnitřní energii lépe představili, podívejme se na ni mikroskopicky. Co přesně tvoří ono "U" v naší nádobě s plynem?

* **Translační kinetická energie molekul:** Molekuly neustále létají sem a tam (čím vyšší teplota, tím rychleji).
* **Rotační a vibrační energie molekul:** Víceatomové molekuly (např. $O_2$ nebo $H_2O$) navíc rotují kolem svých os a atomy v nich vibrují jako na pružinkách.
* **Energie chemických vazeb:** Energie uložená ve vazbách mezi atomy (tato energie se uvolňuje nebo spotřebovává při chemických reakcích, např. při hoření paliva v motoru).
* **Potenciální energie mezimolekulárních sil:** V kapalinách a pevných látkách, ale i v reálných plynech pod vysokým tlakem, se molekuly navzájem přitahují nebo odpuzují.
* **Energie elektronových a jaderných stavů.**

Všechny tyto mikroskopické pohyby a interakce vnímáme z makroskopického hlediska zkrátka jako jedinou termodynamickou vlastnost – vnitřní energii $U$.

## Přenos energie teplem

Zatím jsme se bavili o přenosu energie pomocí **práce** (hřídel se točí, píst stlačuje plyn, teče proud). Uzavřený systém ale může s okolím interagovat ještě jedním klíčovým způsobem, který do definice práce nespadá.

Představte si nádobu s plynem (konstantní objem, žádné pohyblivé části, žádné dráty). Pokud ji položíme na horkou plotnu, teplota plynu vzroste. Jeho vnitřní energie $U$ se tedy zvýšila, ale nebyla vykonána vůbec žádná práce. Tuto interakci nazýváme **přenos energie teplem**.

> Teplo a práce jsou z termodynamického hlediska rovnocenné – obojí představuje energii *v procesu přenosu* přes hranici systému. Nejsou to "látky", které by v systému žily. Systém v sobě nemá "určité množství tepla" nebo "určité množství práce". Má pouze vnitřní energii $U$.

Z historických i experimentálních důvodů (počínaje pracemi J. P. Joulea) víme, že přenos tepla nastává z jediného důvodu: **kvůli rozdílu teplot mezi systémem a okolím**. Teplo vždy samovolně přechází z tělesa o vyšší teplotě na těleso o nižší teplotě.

### Znaménková konvence a značení tepla

Množství energie přenesené teplem značíme symbolem $Q$. Termodynamická konvence je zde (na rozdíl od konvence pro inženýrskou práci) postavena na pohledu ze středu systému:

* **$Q > 0$:** Teplo se dodává do systému (systém přijímá energii, ohřívá se).
* **$Q < 0$:** Teplo se odvádí ze systému (systém ztrácí energii, chladne).

Proces, při kterém nedochází k žádnému přenosu tepla ($Q = 0$), se nazývá **adiabatický děj** (například velmi rychlý děj, nebo děj probíhající v dokonale izolované termosce).

> **Proč to nesouladí se znaménky pro práci?** Inženýrská konvence ($W_{out} > 0, Q_{in} > 0$) vznikla v éře parních strojů. Inženýři chtěli, aby to "dobré", co systém dělá (koná práci a táhne vlak), bylo kladné, a to "nutné", co musí kupovat (teplo z uhlí k ohřevu kotle), bylo také kladné. Pro analýzu motorů a turbín je to zkrátka pohodlnější.

Stejně jako práce, ani teplo nezávisí jen na počátečním a konečném stavu, ale i na způsobu přenosu. **Teplo není vlastnost.** Používáme proto neúplný diferenciál $\delta Q$.

Rychlost přenosu tepla (tepelný výkon) se značí $\dot{Q}$ (v jednotkách W nebo kW). Často pracujeme s tzv. **hustotou tepelného toku** $\dot{q}$, což je tepelný výkon vztažený na jednotkovou plochu ($\text{W/m}^2$). Celkový výkon přes danou plochu $A$ pak zjistíme integrací: $\dot{Q} = \int \dot{q} \, dA$.

### Způsoby šíření tepla

Ačkoliv pro bilanci energie nám stačí vědět "kolik" tepla se přeneslo, inženýrská praxe potřebuje vědět "jak rychle" a "jakým způsobem" se teplo šíří, aby šlo navrhnout např. chladič. Známe tři základní mechanismy sdílení tepla:

#### 1. Vedení (Kondukce)

Teplo se šíří uvnitř pevných látek, kapalin nebo plynů na mikroskopické úrovni předáváním kinetické energie mezi sousedními částicemi bez makroskopického proudění hmoty.
Řídí se **Fourierovým zákonem**. Pro jednorozměrnou stěnu platí, že tepelný tok $\dot{Q}_x$ je úměrný ploše stěny $A$, teplotnímu gradientu (rozdílu teplot $\Delta T$ dělenému tloušťkou $L$) a tepelné vodivosti materiálu $k$ (tzv. Fourierův součinitel tepelné vodivosti).
Tepelná vodivost určuje, zda je materiál dobrý vodič (měď) nebo izolant (polystyren).

#### 2. Záření (Radiace)

Energie se šíří formou elektromagnetického vlnění. Nepotřebuje žádné látkové prostředí a může se šířit i vakuem. Podle **Stefan-Boltzmannova zákona** je energie vyzářená absolutně černým tělesem přímo úměrná čtvrté mocnině jeho absolutní (termodynamické) teploty ($T^4$). Proto radiace hraje extrémně dominantní roli při vysokých teplotách (např. pec, hoření v motoru). Reálné povrchy vyzařují méně než ideální černé těleso, což popisuje vlastnost zvaná emisivita ($\varepsilon$).

#### 3. Proudění (Konvekce)

Nastává mezi pevnou stěnou a proudící tekutinou (plynem nebo kapalinou). Jde o kombinaci mikroskopického vedení těsně u stěny a makroskopického odnášení tepla hmotou proudící tekutiny. Popisuje jej inženýrský empirický vztah zvaný **Newtonův ochlazovací zákon**:


$$ \dot{Q}_c = h A (T_b - T_f) $$


Kde $T_b$ je teplota povrchu, $T_f$ je teplota proudící tekutiny, $A$ je plocha a $h$ je **součinitel přestupu tepla**. Hodnota $h$ velmi závisí na tom, o jakou tekutinu jde a zda je proudění přirozené (vzduch stoupající od horkého radiátoru) nebo nucené (vzduch hnaný přes procesor výkonným ventilátorem).
