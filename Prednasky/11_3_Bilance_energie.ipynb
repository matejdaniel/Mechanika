## Bilance energie pro uzavřené systémy

V předchozích kapitolách jsme si definovali všechny formy, kterými může systém uchovávat energii (kinetická, potenciální, vnitřní), a způsoby, kterými může energie překračovat hranice systému (práce a teplo). Nyní tyto znalosti spojíme do jednoho mocného celku. 

Při inženýrské analýze se často cítíme trochu jako účetní. Musíme pečlivě sledovat, co do systému vchází, co z něj vychází a co v něm zůstává. Základním pravidlem tohoto "účetnictví" je **První termodynamický zákon**, který v podstatě říká, že energie se nedá ani stvořit, ani zničit. Lze ji pouze přeměňovat a přenášet.


### Základní rovnice energetické bilance

Pokud víme, že energie se zachovává a pro uzavřený systém existují jen dva způsoby jejího přenosu (teplo a práce), můžeme první termodynamický zákon formulovat slovně takto:

> Změna celkové energie uvnitř systému za určitý časový úsek se musí rovnat čistému množství energie dodané do systému teplem **mínus** čisté množství energie odevzdané systémem ve formě práce.

Matematicky zapsáno pomocí symbolů získáváme fundamentální rovnici energetické bilance pro uzavřený systém:

$$ E_2 - E_1 = Q - W $$

Při podrobnějším rozpisu změn makroskopických a mikroskopických forem energie (podle rovnice z minulé kapitoly $\Delta E = \Delta E_k + \Delta E_p + \Delta U$) dostáváme:

$$ \Delta E_k + \Delta E_p + \Delta U = Q - W $$

```{warning}
Všimněte si rozdílných znamének před teplem ($Q$) a prací ($W$). Tato asymetrie přímo vyplývá z naší inženýrské znaménkové konvence, kterou jsme přijali již dříve a kterou je klíčové si hlídat:

* Když do systému **vstupuje teplo**, systém získává energii ($Q > 0$, proto má v rovnici kladné znaménko).
* Když systém **koná práci na okolí**, systém odevzdává energii ($W > 0$, proto má v rovnici záporné znaménko, aby se celková energie systému snížila).
```

### Jiné formy bilance energie

V závislosti na tom, co přesně analyzujeme, můžeme rovnici přepsat do dalších užitečných tvarů:

**1. Diferenciální tvar:** (pro analýzu velmi malých změn stavu)


$$ dE = \delta Q - \delta W $$


*(Zde opět připomínáme: $E$ je vlastnost a má úplný diferenciál $dE$, zatímco $Q$ a $W$ vlastnosti nejsou a mají neúplný diferenciál $\delta$).*

**2. Rychlostní (výkonový) tvar:** (pro analýzu dějů v čase)


$$ \frac{dE}{dt} = \dot{Q} - \dot{W} $$


Slovně: Rychlost změny celkové energie v čase se rovná tepelnému výkonu dodávanému do systému bez odevzdávaného mechanického/elektrického výkonu.

**3. Ustálený stav (Steady state):**
Mnoho inženýrských zařízení (motory, převodovky při konstantních otáčkách) pracuje v ustáleném stavu. To znamená, že žádná z jejich vlastností se v čase nemění. V takovém případě je $\frac{dE}{dt} = 0$ a rovnice se zjednoduší na:

$$ \dot{Q} = \dot{W} $$


### Bilance energie pro termodynamické cykly

Jak jsme si uvedli v úvodu do termodynamiky, **termodynamický cyklus** je sled dějů, po jejichž proběhnutí se systém vrátí přesně do svého výchozího stavu.

Jelikož celková energie $E$ je vlastností systému a počáteční a konečný stav cyklu jsou identické, celková změna energie po jednom celém cyklu je přesně nulová ($\Delta E_{cycle} = 0$).

Aplikací energetické bilance na celý cyklus dostáváme:


$$ \Delta E_{cycle} = Q_{cycle} - W_{cycle} = 0 $$

$$ W_{cycle} = Q_{cycle} $$

Tento vztah platí pro naprosto každý termodynamický cyklus bez ohledu na látku v systému nebo složitost procesu. Čistá práce odevzdaná za jeden cyklus se musí rovnat čistému teplu dodanému za tento cyklus.


## Princip tepelného čerpadla

Abychom si celé toto energetické "účetnictví" ukázali v akci, rozebereme si jeden z inženýrských vynálezů, který se dnes masivně používá k vytápění domů: **tepelné čerpadlo** (Heat pump).

Zkušenost i příroda nám říkají, že teplo teče vždy z teplejšího tělesa na chladnější. Jak je tedy možné vytápět dům (kde je +20 °C) z venkovního vzduchu (kde je např. -5 °C)? Musíme energii "pumpovat" proti jejímu přirozenému směru – a k tomu potřebujeme vynaložit práci.

```{image} https://www.energy.gov/sites/default/files/styles/full_article_width/public/2021-11/shutterstock_622693838%20%5BConverted%5D.jpg?itok=ZJAFO7pW
```

### Energetická bilance tepelného čerpadla

Představme si náš systém jako chladivo obíhající v trubkách tepelného čerpadla. Pro zjednodušení zápisu se zde (na rozdíl od obecné konvence) často zavádějí absolutní hodnoty toků definované specificky pro tyto stroje:

1. **$Q_{in}$ (Teplo z chladného tělesa):** Chladivo přijímá teplo z venkovního prostředí (vzduchu, země, vody). I když je venku zima, stále je tam obrovské množství tepelné energie.
2. **$W_{cycle}$ (Práce dodaná do systému):** Kompresor, poháněný elektřinou ze sítě, stlačí chladivo. Tím vykoná na systému práci.
3. **$Q_{out}$ (Teplo odevzdané do teplého tělesa):** Stlačené a zahřáté chladivo odevzdá teplo do topné soustavy domu.

Z hlediska energetické bilance prvního zákona pro cyklus musí platit, že všechna energie, která do systému vstoupí, ho musí také opustit:

$$ \text{Energie dovnitř} = \text{Energie ven} $$

$$ Q_{in} + W_{cycle} = Q_{out} $$

Pokud tuto rovnici přepíšeme do tvaru pro potřebnou dodanou práci, získáme:


$$ W_{cycle} = Q_{out} - Q_{in} $$

### Hodnocení výkonnosti: Topný faktor (COP)

Když si kupujete topidlo, zajímá vás, kolik užitečného tepla dostanete za peníze (elektrickou práci), které zaplatíte. Výkonnost tepelných čerpadel a chladicích zařízení se proto vyjadřuje pomocí **koeficientu výkonnosti** (Coefficient of Performance). Pro tepelné čerpadlo, kde je naším užitečným cílem vytápět ($Q_{out}$), definujeme topný faktor $\gamma$ (často označován jako $COP_{HP}$) takto:

$$ \gamma = \frac{\text{Užitečný efekt}}{\text{Vynaložené náklady}} = \frac{Q_{out}}{W_{cycle}} $$

Pokud do rovnice dosadíme naši energetickou bilanci ($W_{cycle} = Q_{out} - Q_{in}$), dostaneme:

$$ \gamma = \frac{Q_{out}}{Q_{out} - Q_{in}} $$

**Proč jsou tepelná čerpadla tak výhodná?**
Když doma zapnete běžný elektrický přímotop, všechna elektřina se jen přemění na teplo ($W \rightarrow Q_{out}$). Získáte přesně tolik tepla, kolik zaplatíte na elektřině (účinnost 100 %, tedy poměr je 1).

Podívejte se ale na rovnici energetické bilance tepelného čerpadla ($Q_{out} = Q_{in} + W_{cycle}$). Teplo odevzdané do domu je tvořeno prací z kompresoru *PLUS* teplem, které jsme zdarma "vysáli" zvenku! V důsledku toho je topný faktor $\gamma$ **vždy větší než 1**. Běžná tepelná čerpadla mají COP kolem 3 až 4. To znamená, že za každý 1 kW spotřebované elektrické práce získáte 3 až 4 kW tepla pro váš dům. Nejde o porušení zákona zachování energie, pouze o chytré využití energie z okolí, o čemž nás přesvědčuje právě energetická bilance.
