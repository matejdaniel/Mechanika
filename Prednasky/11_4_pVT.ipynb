## Základy hodnocení vlastností a termodynamické diagramy

Abychom mohli aplikovat naši rovnici energetické bilance z předchozí kapitoly, musíme znát konkrétní hodnoty termodynamických vlastností ($u$, $h$, $v$, atd.) ve výchozím a konečném stavu systému.

### Určování stavu (Stavový princip)

Jak jsme se již dozvěděli, **stav** uzavřeného systému v rovnováze je plně popsán jeho termodynamickými vlastnostmi. Z pozorování víme, že tyto vlastnosti na sobě nejsou nezávislé. Pokud jich známe určitý počet, všechny ostatní už z nich lze jednoznačně vypočítat.

Otázka zní: *Kolik nezávislých vlastností potřebuji znát, abych jednoznačně určil stav systému?*

Odpověď dává **Stavový princip (State principle)**. Ten vychází ze zkušenosti a říká: *Počet nezávislých vlastností se rovná jedné (za tepelný přenos) plus počet relevantních způsobů konání kvazistatické práce.*

Většina našich inženýrských systémů (plyny, pára, chladiva) se dá modelovat jako **jednoduchý stlačitelný systém**. To znamená, že jediný relevantní způsob konání práce je změna objemu ($\int p\,dV$). Pro takové systémy stavový princip říká, že **stav je plně určen dvěma nezávislými intenzivními vlastnostmi**.

> **Příklad:** Pokud máme v pístu páru a změříme její tlak ($p$) a teplotu ($T$), její stav je plně fixován. Specifický objem ($v$) i měrná vnitřní energie ($u$) mají v tento moment jedinou možnou hodnotu: $v = f(p, T)$ a $u = g(p, T)$.


### p-v-T vztah a stavový povrch

Pokud vezmeme tři nejběžnější měřitelné vlastnosti – tlak ($p$), měrný objem ($v$) a teplotu ($T$) – můžeme jejich vztah pro jednoduchou látku vynést do trojrozměrného grafu. Výsledkem je tzv. **p-v-T povrch**. Každý bod na tomto povrchu představuje rovnovážný stav látky.

Tento povrch ukazuje komplexní chování látky ve všech jejích fázích (pevná látka, kapalina, plyn/pára) a při přechodech mezi nimi (tání, var, sublimace).

Ačkoliv je 3D povrch užitečný pro pochopení celku, v inženýrské praxi je mnohem praktičtější pracovat s jeho 2D průměty.

#### Pracovní diagram (p-v diagram)

Pokud povrch promítneme do roviny tlak-měrný objem ($p-v$), získáme **pracovní diagram**. Tento diagram jsme již používali u výpočtu objemové práce (plocha pod křivkou).
Na tomto diagramu je jasně vidět zvonovitá křivka oddělující dvoufázovou oblast (kapalina-pára). Této oblasti se říká **zvon sytosti** a vrcholí v kritickém bodě. Čáry konstantní teploty (izotermy) na tomto diagramu uvnitř zvonu sytosti probíhají vodorovně (var probíhá při konstantním tlaku i teplotě).

```{image} https://pressbooks.bccampus.ca/thermo1/wp-content/uploads/sites/499/2021/06/Fig-2-9_P-v_diagram_for_a_liquid-vapor.svg_.png
```

#### Teplotní diagram (T-v diagram)

Podobně jako p-v diagram, i T-v diagram se často používá k zakreslování procesů. Místo tlaku je na svislé ose teplota. Čáry konstantního tlaku (izobary) zde uvnitř zvonu sytosti také probíhají vodorovně.

```{image} https://pressbooks.bccampus.ca/thermo1/wp-content/uploads/sites/499/2021/06/Fig.-2-8_T-v_diagram_for_a_liquid_vapor-copy.svg_-768x615.png
```

### Pochopení fázové změny (varu) na diagramu

Zkusme si rozebrat jednoduchý proces ohřevu vody při konstantním tlaku (např. 1 atmosféra) a sledovat jej na $T-v$ diagramu.

1. **Podchlazená kapalina (Subcooled / Compressed liquid):** Začínáme s vodou při 20 °C. Jsme nalevo od zvonu sytosti. Jak dodáváme teplo, teplota roste, objem roste jen velmi málo.
2. **Sytá kapalina (Saturated liquid):** Dosáhneme 100 °C (pro 1 atm). Jsme přesně na levé hraně zvonu sytosti. Voda je těsně před varem. (Index vlastností se často značí **$f$**).
3. **Mokrá pára (Dvoufázová oblast):** Dodáváme teplo, ale teplota se nemění, zůstává 100 °C. Kapalina se mění na páru, objem prudce roste. Pohybujeme se vodorovně uvnitř zvonu. Zde je nutné definovat **suchost páry (Quality) $x$**, což je hmotnostní podíl páry ve směsi ($x = \frac{m_{pary}}{m_{celkem}}$). Platí, že $x$ je od 0 (sytá kapalina) do 1 (sytá pára).
4. **Sytá pára (Saturated vapor):** Poslední kapka se vypařila. Jsme na pravé hraně zvonu. (Index vlastností se často značí **$g$**).
5. **Přehřátá pára (Superheated vapor):** Pokud dál topíme, jsme napravo od zvonu. Teplota opět začíná prudce stoupat a objem roste.

> Všechny termodynamické vlastnosti ($v$, $u$, $h$) mokré páry uvnitř zvonu se počítají jako vážený průměr syté kapaliny a syté páry pomocí suchosti $x$:
> 
> $$ v = v_f + x(v_g - v_f) $$
