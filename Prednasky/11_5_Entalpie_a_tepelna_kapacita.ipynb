## Entalpie, měrná tepelná kapacita a vlastnosti kapalin

V předchozí kapitole jsme analyzovali tepelné čerpadlo jako jeden velký černý box pomocí celkového tepla a práce za celý cyklus. V inženýrské praxi ale potřebujeme analyzovat každý komponent zvlášť – kompresor, kondenzátor, expanzní ventil a výparník. Těmito zařízeními neustále protéká chladivo. Pro popis takto proudící tekutiny musíme zavést novou termodynamickou vlastnost.

### Co je to entalpie?

**Entalpie** (značí se $H$, pro 1 kg látky používáme měrnou entalpii $h$) je termodynamická vlastnost (stavová veličina), která vyjadřuje celkovou tepelnou a tlakovou energii, kterou v sobě látka uchovává.

Matematicky je entalpie definována jako součet vnitřní energie systému a součinu jeho tlaku a objemu:

$$ H = U + pV $$

Pro 1 kg látky (měrná entalpie v kJ/kg) píšeme:

$$ h = u + pv $$

Kde:

* **$U$ (nebo $u$)** je vnitřní energie látky (pohyb a interakce molekul).
* **$p$** je tlak.
* **$V$ (nebo $v$)** je objem (nebo měrný objem).

* Člen **$pV$** představuje tzv. **práci protlačení** (flow work). Je to mechanická energie potřebná k tomu, aby tekutina vůbec mohla překonat okolní tlak, zaujmout svůj objem a proudit systémem.

### Inženýrský význam entalpie

Proč si inženýři vymysleli entalpii, když už mají vnitřní energii? Hlavním důvodem je **extrémní zjednodušení výpočtů u otevřených systémů** (zařízení, kterými neustále protéká tekutina).

Když plyn nebo kapalina protéká strojem, nese si s sebou nejen svou vnitřní energii ($u$), ale okolní tekutina na ni musí neustále tlačit silou, aby ji do stroje natlačila a následně vytlačila ven (práce protlačení $pv$). Protože se tyto dva členy ($u$ a $pv$) u proudících tekutin vyskytují v energetické bilanci *vždy společně*, inženýři je spojili do jedné veličiny – entalpie ($h$).

Zde jsou tři hlavní oblasti, kde je entalpie naprosto klíčová:

1. **Analýza otevřených systémů** (turbíny, kompresory, čerpadla)

V zařízeních pracujících v ustáleném stavu se zákon zachování energie (první termodynamický zákon) díky entalpii redukuje na velmi jednoduché vztahy. Pokud zanedbáme změny kinetické a potenciální energie, platí:

* **Kompresor/Čerpadlo:** Práce, kterou musíte kompresoru dodat, je přesně rovna rozdílu entalpií na výstupu a vstupu ($w = h_{vystup} - h_{vstup}$).
* **Turbína:** Práce, kterou turbína vyrobí, je rovna úbytku entalpie proudící páry ($w = h_{vstup} - h_{vystup}$).

2. **Tepelné výměníky a izobarické děje** (bojlery, chladiče)

Většina ohřevů a chlazení v průmyslu probíhá v trubkách nebo nádobách při konstantním tlaku (izobarický děj). Za těchto podmínek platí úžasné pravidlo: **Teplo dodané do systému (nebo z něj odebrané) se rovná přesně změně jeho entalpie.**

* V parním kotli elektrárny dodáváte teplo $q = h_{para} - h_{voda}$. Nemusíte složitě počítat, jaká část tepla šla na zvýšení teploty a jaká na expanzní práci – entalpie to řeší za vás.

3. **Fázové přeměny** (Var a kondenzace)

Když voda vaří a mění se na páru, potřebuje přijmout obrovské množství energie (skupenské teplo varu). V termodynamických tabulkách se tato hodnota nachází jako rozdíl entalpie syté páry a syté kapaliny ($h_{fg} = h_g - h_f$). V inženýrství se tento údaj používá při návrhu chladicích cyklů, klimatizací i tepelných čerpadel.

**Shrnutí:** Zatímco vnitřní energie ($U$) je skvělá pro analýzu zavřené nádoby (např. plyn uzavřený v pevném tlakovém hrnci), **entalpie ($H$)** je primárním nástrojem inženýra pro všechno, co proudí trubkami a stroji. Zjednodušuje rovnice a umožňuje snadné čtení energetických změn z termodynamických tabulek.

### Entalpie na příkladu tepelného čerpadla

Představte si kompresor tepelného čerpadla. Chladivo (plyn) do něj vstupuje určitou rychlostí, nese si svou vnitřní energii $u$ a kompresor na něj koná práci, aby ho stlačil a vytlačil ven do kondenzátoru. Aby tekutina vůbec mohla vstoupit do kontrolního objemu kompresoru a následně ho opustit, musí překonat tlak okolní tekutiny. K tomu je potřeba tzv. **práce protlačení** (flow work), která je matematicky rovna součinu tlaku a měrného objemu ($p \cdot v$).

Celková termodynamická energie, kterou s sebou proudící 1 kg chladiva nese, je tedy součtem jeho vnitřní energie a práce protlačení. Tento součet se v termodynamických výpočtech objevuje tak často, že dostal vlastní název a symbol – **Entalpie ($h$)**.

Z definičního vztahu platí pro měrnou entalpii:

$$ h = u + pv $$

#### Určení vlastností v cyklu tepelného čerpadla

Díky entalpii se energetická bilance jednotlivých komponent extrémně zjednoduší. Všechna vyměněná energie se dá vyjádřit jako pouhý rozdíl entalpií na vstupu a výstupu!

Ukažme si to na 4 základních krocích tepelného čerpadla:

1. **Kompresor:** Přijímá chladnou páru z výparníku (stav 1) a stlačuje ji na horký plyn (stav 2). Práce, kterou musíme kompresoru dodat na 1 kg chladiva, je jednoduše $w_{comp} = h_2 - h_1$.
2. **Kondenzátor:** Horký plyn (stav 2) kondenzuje na kapalinu (stav 3) a odevzdává užitečné teplo do domu. Předané teplo je $q_{out} = h_2 - h_3$.
3. **Expanzní ventil:** Kapalina (stav 3) protéká úzkým ventilem, kde prudce klesne její tlak (stav 4). Tento proces je tak rychlý a bez výměny tepla a práce, že entalpie se nemění (izentalpický děj). Platí $h_4 = h_3$.
4. **Výparník:** Chladná směs (stav 4) se odpařuje tím, že odebírá teplo z venkovního vzduchu a mění se zpět na páru (stav 1). Přijaté teplo je $q_{in} = h_1 - h_4$.

Topný faktor (COP) pak můžeme velmi elegantně spočítat pouze ze znalosti entalpií v těchto bodech (které vyčteme z termodynamických tabulek na základě teploty a tlaku):


$$ COP = \frac{q_{out}}{w_{comp}} = \frac{h_2 - h_3}{h_2 - h_1} $$


### Měrná tepelná kapacita ($c_v$ a $c_p$)

Dalšími nesmírně důležitými vlastnostmi, které úzce souvisí s vnitřní energií a entalpií, jsou **měrné tepelné kapacity**. Ty nám říkají, jak citlivá je teplota látky na dodání tepelné energie.

Pro čisté, jednoduché stlačitelné látky definujeme dvě měrné tepelné kapacity pomocí parciálních derivací:

* **Měrná tepelná kapacita při stálém objemu ($c_v$):**

$$ c_v = \left( \frac{\partial u}{\partial T} \right)_v $$


* **Měrná tepelná kapacita při stálém tlaku ($c_p$):**

$$ c_p = \left( \frac{\partial h}{\partial T} \right)_p $$

Jednotkou v soustavě SI je $\text{kJ/(kg} \cdot \text{K)}$. Fyzikální význam je zřejmý: udávají množství energie potřebné k ohřátí 1 kg látky o 1 Kelvin (při izochorickém, respektive izobarickém ději). Dále se často zavádí **Poissonova konstanta** (někdy označována $k$ nebo $\kappa$), což je bezrozměrný poměr těchto dvou kapacit:

$$ k = \frac{c_p}{c_v} $$

U plynů závisí $c_p$ i $c_v$ nejen na teplotě, ale částečně i na tlaku (obzvlášť v blízkosti syté páry).


### Hodnocení vlastností kapalin a pevných látek (Model nestlačitelné látky)

Zatímco pro vodní páru nebo chladiva potřebujeme podrobné termodynamické tabulky, pro kapaliny (jako je tekoucí voda v potrubí) a pevné látky si inženýři situaci často zjednodušují.

Z tabulek podchlazené kapaliny víme, že měrný objem $v$ a vnitřní energie $u$ kapalin se s tlakem mění jen velmi nepatrně. Jsou silně závislé téměř výhradně na teplotě. Můžeme je proto s velkou přesností aproximovat hodnotami pro sytou kapalinu při dané teplotě:


$$ v(T, p) \approx v_f(T) $$

$$ u(T, p) \approx u_f(T) $$

Tato zkušenost nás vede k vytvoření **modelu nestlačitelné látky**. U tohoto modelu předpokládáme dvě věci:

1. Měrný objem (hustota) je absolutně konstantní.
2. Vnitřní energie závisí pouze na teplotě ($u = u(T)$).

Z tohoto modelu vyplývá jeden zásadní důsledek pro měrné tepelné kapacity. Protože kapaliny a pevné látky při ohřívání prakticky nemění objem (nekonají objemovou práci), nezáleží na tom, zda je ohříváme při stálém objemu nebo tlaku. Platí tedy:

$$ c_p = c_v = c $$

Pro nestlačitelnou látku máme pouze jednu měrnou tepelnou kapacitu $c$. Změny vnitřní energie a entalpie (pokud považujeme $c$ za konstantní v daném teplotním rozsahu) pak spočítáme analyticky:

$$ \Delta u = u_2 - u_1 = c(T_2 - T_1) $$

$$ \Delta h = h_2 - h_1 = c(T_2 - T_1) + v(p_2 - p_1) $$

Člen $v(p_2 - p_1)$ v rovnici pro entalpii představuje vliv změny tlaku. Pokud tekutinu ohříváme čerpadlem nebo se její tlak příliš nemění, je tento člen často zanedbatelný a platí jednoduše $\Delta h \approx c \Delta T$.
