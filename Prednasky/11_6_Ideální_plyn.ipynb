## Ideální plyn

Ideální plyn je teoretický (hypotetický) plyn, který splňuje dvě základní podmínky, jež mikroskopicky vycházejí z kinetické teorie plynů:

1. **Žádné mezimolekulární síly:** Molekuly plynu na sebe nijak nepůsobí (nepřitahují se ani neodpuzují), s výjimkou dokonale pružných srážek.
2. **Zanedbatelný vlastní objem:** Samotné molekuly jsou považovány za bodové částice, jejichž vlastní objem je nulový ve srovnání s celkovým objemem nádoby.

Z makroskopického hlediska je ideální plyn definován tak, že přesně splňuje **stavovou rovnici ideálního plynu**:

$$ pv = RT $$

Kde:

* $p$ je absolutní tlak.
* $v$ je měrný objem (objem na jednotku hmotnosti, $v = V/m$).
* $T$ je absolutní termodynamická teplota (v Kelvinech).
* $R$ je měrná plynová konstanta pro daný plyn ($R = R_u / M$, kde $R_u$ je univerzální plynová konstanta a $M$ je molární hmotnost plynu).

Rovnici lze zapsat i pomocí celkového objemu $V$ a hmotnosti $m$:

$$ pV = mRT $$

**Kdy můžeme tento model použít?**
Pro drtivou většinu inženýrských aplikací (např. analýza spalovacích motorů, plynových turbín nebo klimatizací) se chování běžných plynů jako je vzduch, dusík, kyslík, vodík či spaliny při atmosférickém tlaku a běžných teplotách téměř dokonale shoduje s modelem ideálního plynu. Vodní pára v parní turbíně se naopak jako ideální plyn chovat *nesmí* a musíme pro ni použít parní tabulky.


### Vnitřní energie ideálního plynu

Abychom mohli odvodit entalpii, musíme se nejprve podívat na vnitřní energii $u$.

Z mikroskopického pohledu tvoří vnitřní energii plynu kinetická energie molekul (translační, rotační, vibrační) a potenciální energie jejich vzájemného silového působení.

Protože jsme si v definici ideálního plynu řekli, že **mezimolekulární síly neexistují**, je potenciální energie mezi molekulami přesně nulová. Vnitřní energie ideálního plynu je tak tvořena čistě jen kinetickou energií molekul.

Kinetická energie částic závisí *pouze a jen na teplotě*. Z toho vyplývá naprosto zásadní termodynamický závěr: **Vnitřní energie ideálního plynu je funkcí pouze teploty.** Nezávisí na tlaku ani na objemu.

Matematicky zapsáno:

$$ u = u(T) $$

Díky tomu se definice měrné tepelné kapacity při stálém objemu ($c_v = \frac{\partial u}{\partial T}$) pro ideální plyn mění z parciální derivace na obyčejnou derivaci:

$$ c_v(T) = \frac{du}{dT} \implies du = c_v(T) \, dT $$


### Odvození entalpie ideálního plynu

Nyní máme vše připraveno k tomu, abychom odvodili tvar entalpie pro ideální plyn. Postupujeme přesně podle její obecné definice:

$$ h = u + pv $$

Z modelu ideálního plynu víme, že pro člen práce protlačení platí $pv = RT$. Tento vztah jednoduše dosadíme do definice entalpie:

$$ h = u + RT $$

Podívejme se na pravou stranu výsledné rovnice:

* Víme, že $u$ je funkcí pouze teploty: $u = u(T)$.
* $R$ je pevná konstanta pro daný plyn.
* $T$ je teplota.

Celá pravá strana rovnice tedy nezávisí na ničem jiném než na teplotě $T$. Z toho logicky vyplývá, že **entalpie ideálního plynu závisí také pouze na teplotě**. Nezávisí na tlaku ani na objemu.

$$ h = h(T) $$

Tento výsledek nám opět extrémně zjednodušuje definici měrné tepelné kapacity při stálém tlaku ($c_p = \frac{\partial h}{\partial T}$), která se mění na obyčejnou derivaci:

$$ c_p(T) = \frac{dh}{dT} \implies dh = c_p(T) \, dT $$

Změnu entalpie ideálního plynu tedy můžeme vždy vypočítat integrací kapacity $c_p$ podle teploty: $\Delta h = \int c_p \, dT$.


### Mayerův vztah (Vztah mezi $c_p$ a $c_v$)

Odvození entalpie ideálního plynu nám otevírá cestu k velmi elegantnímu zjištění, jaký je přesný vztah mezi měrnými tepelnými kapacitami $c_p$ a $c_v$.

Vezměme si naši odvozenou rovnici pro entalpii ideálního plynu:

$$ h = u + RT $$

Zderivujme celou tuto rovnici podle teploty $T$:


$$ \frac{dh}{dT} = \frac{du}{dT} + \frac{d(RT)}{dT} $$

Protože $R$ je konstanta, derivace termínu $RT$ podle $T$ je jednoduše $R$:

$$ \frac{dh}{dT} = \frac{du}{dT} + R $$

Nyní do rovnice dosadíme definice měrných tepelných kapacit, které jsme si ukázali výše ($\frac{dh}{dT} = c_p$ a $\frac{du}{dT} = c_v$):

$$ c_p = c_v + R $$

Případně ve známějším tvaru (tzv. **Mayerův vztah**):

$$ c_p - c_v = R $$

Tento vztah je neuvěřitelně užitečný. Říká nám, že pro ideální plyn je měrná tepelná kapacita při stálém tlaku ($c_p$) vždy větší než při stálém objemu ($c_v$) přesně o hodnotu plynové konstanty $R$. K ohřátí plynu o 1 stupeň při konstantním tlaku potřebujete více energie než při konstantním objemu, protože plyn musí navíc vykonat expanzní práci ($pv = RT$).

## Popis základních dějů 

V termodynamice často analyzujeme děje, při kterých zůstává určitá stavová veličina konstantní, nebo je potlačen určitý typ interakce s okolím. Pochopení těchto základních dějů je klíčové pro analýzu složitějších cyklů (jako je Carnotův, Ottův nebo Dieselův cyklus).

Zde jsou definice tří základních termodynamických dějů:

### Izotermický děj (Stálá teplota)

Při izotermickém ději zůstává **teplota systému konstantní ($T = \text{konst.}$)** po celou dobu trvání procesu. Aby teplota zůstala stálá při expanzi nebo kompresi, musí systém neustále vyměňovat teplo se svým okolím (např. být ponořen do velké tepelné lázně). Tento děj obvykle probíhá velmi pomalu.

* **Chování ideálního plynu:** Protože vnitřní energie $u$ i entalpie $h$ ideálního plynu závisí pouze na teplotě, při izotermickém ději se **nemění** ($\Delta U = 0$, $\Delta H = 0$).
* **Energetická bilance ($Q - W = \Delta U$):** Z rovnice vyplývá, že $Q = W$. Veškeré teplo dodané do systému se přemění na objemovou práci (při expanzi), nebo naopak (při kompresi musíme teplo odvádět, jinak by se plyn zahřál).
* **Stavová rovnice:** Z rovnice $pV = mRT$ vyplývá **Boyle-Mariottův zákon**:

$$ pV = \text{konst.} $$



V p-v diagramu se tento děj zobrazuje jako křivka zvaná *izoterma* (matematicky jde o hyperbolu).

### Adiabatický děj (Žádný přenos tepla)

Při adiabatickém ději je systém dokonale tepelně izolován, takže nedochází k **žádné výměně tepla s okolím ($Q = 0$)**. V praxi se adiabatickému ději blíží procesy, které probíhají tak rychle, že systém "nestihne" žádné teplo s okolím vyměnit (např. rychlá expanze horkých plynů v turbíně nebo stlačení směsi ve válci motoru).

* **Energetická bilance ($Q - W = \Delta U$):** Protože $Q = 0$, platí $\Delta U = -W$.
* Pokud plyn expanduje (koná práci, $W > 0$), děje se tak čistě na úkor jeho vnitřní energie ($\Delta U < 0$), a proto se plyn **výrazně ochladí**.
* Pokud plyn stlačujeme ($W < 0$), jeho vnitřní energie a teplota prudce rostou (princip vznícení nafty v dieselovém motoru).


* **Stavová rovnice:** Pro vratný (kvazistatický) adiabatický děj ideálního plynu platí **Poissonova rovnice**:

$$ pV^\kappa = \text{konst.} $$



Kde $\kappa$ (nebo $k$) je Poissonova konstanta ($\kappa = c_p/c_v$). V p-v diagramu se tato křivka nazývá *adiabata* a klesá strměji než izoterma.

### Izochorický děj (Stálý objem)

Při izochorickém ději zůstává **objem systému konstantní ($V = \text{konst.}$)**. Příkladem je ohřev nebo chlazení plynu v pevné, uzavřené tlakové nádobě.

* **Práce:** Protože nedochází ke změně objemu ($dV = 0$), systém nekoná žádnou objemovou práci:

$$ W = \int p \, dV = 0 $$


* **Energetická bilance ($Q - W = \Delta U$):** Protože $W = 0$, rovnice se zjednoduší na $Q = \Delta U$. Veškeré dodané teplo se spotřebuje výhradně na zvýšení vnitřní energie (a tedy i teploty a tlaku) plynu.
* **Stavová rovnice:** Z rovnice ideálního plynu vyplývá **Charlesův zákon**:

$$ \frac{p}{T} = \text{konst.} $$



Tlak plynu roste přímo úměrně s jeho absolutní teplotou. V p-v diagramu se tento děj zobrazuje jako svislá úsečka.

*(Kromě těchto tří existuje samozřejmě i děj **izobarický**, při kterém zůstává konstantní tlak, $p = \text{konst.}$, a který se v p-v diagramu zobrazuje jako vodorovná úsečka. Ten jsme si podrobně popsali již v kapitole o objemové práci a entalpii).*
