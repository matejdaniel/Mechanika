## Druhý termodynamický zákon a Carnotův cyklus

Zatím jsme se v našich analýzách opírali o princip zachování energie, tedy první termodynamický zákon. Ten je nezbytný, ale ukazuje se, že ne vždy postačuje. Náš každodenní život je plný dějů, které se přirozeně odehrávají pouze jedním směrem. Šálek horké kávy v místnosti vždy zchladne. Zmrzlina v létě vždy roztaje. Pokud bychom chtěli, aby to bylo naopak (studená káva se sama od sebe ohřála z okolního vzduchu a vzduch by se při tom ještě ochladil), první termodynamický zákon by s tím neměl problém – energie by se zachovala. Přesto se to nikdy nestane.

Na rozdíl od prvního zákona, který je jen účetním vyrovnáním energie bez ohledu na její kvalitu či směr, **druhý termodynamický zákon určuje právě tento směr přirozených procesů a limity přeměny tepla na práci**.

### Nevratné děje (Irreversibilities)

V předchozím textu jsme se naučili, že přirozené procesy jsou **nevratné (irreversible)**. Jakmile proběhnou, není možné, aby se systém a zároveň jeho okolí vrátily přesně do svého původního stavu bez dodatečného vnějšího zásahu (typicky dodání práce odjinud).

Co konkrétně činí děje nevratnými? Inženýři tyto příčiny označují jako **nevratnosti (irreversibilities)**. Patří mezi ně zejména:

* **Přenos tepla přes konečný teplotní rozdíl:** Teplo proudící z tělesa $T_1$ na $T_2$ (kde $T_1 > T_2$) nelze "zatlačit" zpět, aniž bychom použili chladicí zařízení, které spotřebovává práci.
* **Volná expanze:** Plyn unikající z tlakové nádoby do okolního prostředí se sám nikdy do nádoby nevrátí.
* **Tření:** Jakékoliv mechanické tření nebo viskózní tření v proudící tekutině mění mechanickou energii na "neužitečné" vnitřní teplo.
* **Elektrický odpor, nevratné chemické reakce, míchání plynů, atd.**

Děj, který je teoreticky zcela prostý jakýchkoliv těchto vlivů (v systému i v okolí), se nazývá **vratný děj (reversible process)**. Jde o nedosažitelný ideál (vyžadoval by např. nulové tření nebo nulový rozdíl teplot pro přenos tepla), ale v termodynamice slouží jako absolutní horní limit, se kterým porovnáváme reálné stroje. Často pracujeme s pojmem **vnitřně vratný děj**, což znamená, že nevratnosti se nevyskytují *uvnitř* systému (turbíny či válce), i když v okolí existovat mohou.

### Formulace druhého termodynamického zákona

Protože druhý zákon pokrývá tolik aspektů přírody, existuje několik jeho formulací, které se zdají být odlišné, ale ve skutečnosti jsou ekvivalentní (porušení jednoho znamená porušení druhého).

### Clausiova formulace

> *Není možné sestrojit cyklicky pracující zařízení, jehož jediným výsledkem by byl přenos tepla z chladnějšího tělesa na těleso teplejší.*

Tato formulace přímo zakazuje samovolný tok tepla proti teplotnímu gradientu. Ledničky to samozřejmě umí, ale jejich "jediným výsledkem" není jen přenos tepla – vyžadují také dodání práce ($W_{cycle} \neq 0$) ze sítě do kompresoru.

### Kelvin-Planckova formulace

Pro tuto formulaci si zavedeme pojem **tepelná lázeň (Thermal reservoir)**. Je to teoretický systém, který je tak velký (např. oceán nebo atmosféra), že mu můžete dodávat nebo z něj odebírat teplo, a jeho teplota se nezmění.

> *Není možné sestrojit cyklicky pracující zařízení, které by nepřijímalo teplo z jediné tepelné lázně a přeměnilo ho beze zbytku na ekvivalentní množství práce.*

Tato věta v zásadě říká, že neexistuje stroj se 100% účinností přeměny tepla na práci. Každý cyklický tepelný motor, aby mohl po expanzi fungovat znovu, musí část nevyužitého tepla odevzdat do chladnější lázně (do chladiče, do řeky, do vzduchu). Z toho pramení nutnost, že každý motor musí mít kromě ohřívače i chladič.

### Důsledky a limity cyklů

Z Kelvin-Planckovy formulace plynou dva klíčové důsledky pro jakékoliv tepelné cykly pracující mezi dvěma tepelnými lázněmi (jednou horkou o teplotě $T_H$ a druhou chladnou o teplotě $T_C$):

1. Účinnost nevratného cyklu je vždy menší než účinnost vratného cyklu pracujícího mezi stejnými teplotami.
2. **Všechny vratné cykly pracující mezi stejnými teplotami $T_H$ a $T_C$ mají naprosto stejnou účinnost.**

Tento druhý bod je fascinující. Znamená to, že maximální možná účinnost jakéhokoliv tepelného stroje nezávisí na tom, zda v něm obíhá voda, helium nebo vzduch, ani na typu mechanismu. Závisí **pouze a jedině na rozdílu teplot $T_H$ a $T_C$**.

Na základě tohoto zjištění byla definována termodynamická teplota (Kelvinova stupnice), a díky ní můžeme maximální teoretickou (Carnotovu) účinnost vypočítat prostým vzorcem:

**Maximální účinnost tepelného motoru:**

$$ \eta_{max} = 1 - \frac{T_C}{T_H} $$

*(Teploty musí být dosazeny v Kelvinech!)*

Podobné limity existují i pro chladicí cykly a tepelná čerpadla (opět s teplotami v Kelvinech):

**Maximální topný faktor tepelného čerpadla:**

$$ \gamma_{max} = \frac{T_H}{T_H - T_C} $$

**Maximální topný faktor chladicího cyklu:**

$$ \beta_{max} = \frac{T_C}{T_H - T_C} $$

### Carnotův cyklus

Abychom si dokázali představit onen nedosažitelný "vratný cyklus", který má účinnost rovnou teoretickému maximu, francouzský inženýr Sadi Carnot navrhl myšlenkový model, který nese jeho jméno.

**Carnotův cyklus** se skládá přesně ze čtyř vnitřně vratných dějů, které na sebe navazují. Lze si ho představit buď s plynem v pístu, nebo v systému parní elektrárny:

1. **Vratná izotermická expanze ($2 \to 3$):** Systém přijímá teplo $Q_H$ z horké lázně. Aby děj byl vratný, teplota plynu musí být jen infinitezimálně nižší než teplota lázně, takže se plyn rozpíná a koná práci při konstantní teplotě $T_H$.
2. **Vratná adiabatická expanze ($3 \to 4$):** Nádobu odstavíme od tepla a tepelně izolujeme. Plyn dál expanduje a koná práci, dokud jeho teplota neklesne přesně na teplotu chladné lázně $T_C$.
3. **Vratná izotermická komprese ($4 \to 1$):** Systém kontaktuje chladnou lázeň a my dodáváme práci na stlačení plynu. Tím vzniká teplo $Q_C$, které se odevzdává do lázně při stálé teplotě $T_C$.
4. **Vratná adiabatická komprese ($1 \to 2$):** Systém opět izolujeme a stlačíme ho tak, aby se plyn zahřál zpět přesně na teplotní hladinu $T_H$. Tím je cyklus uzavřen.

Reálně tento cyklus nelze postavit. Izotermický přenos tepla je tak pomalý, že by motor s Carnotovým cyklem neměl téměř žádný výkon. I přes to je to ale zcela nenahraditelný etalon. Inženýr díky rovnici Carnotovy účinnosti hned ví: "Pokud mám v uhelné elektrárně páru o teplotě 500 °C ($T_H$) a chladím ji řekou o teplotě 20 °C ($T_C$), nikdy, ani s dokonalými stroji budoucnosti, nemohu dosáhnout účinnosti lepší než 62 %."

```




### Co tvoří počet mikrostavů ($\Omega$) u ideálního plynu?

Spojení makroskopické termodynamiky (kde $dS = \delta Q / T$) a statistické fyziky (kde $S = k_B \ln \Omega$) je jedním z nejelegantnějších momentů celé fyziky. Podívat se na Carnotův cyklus skrze Boltzmannovu rovnici znamená vyměnit teploměry a tlakoměry za počítání všech možných způsobů, jak se mohou molekuly v systému uspořádat.
Pojďme si Carnotův cyklus rozebrat krok za krokem pod "mikroskopem" statistické mechaniky.

Než spustíme cyklus, musíme si uvědomit, co přesně tvoří onen počet mikrostavů $\Omega$ (statistickou váhu). U ideálního plynu určují mikrostav každé molekuly dvě věci:

1. **Poloha (Prostorové mikrostavy):** Jsou dány celkovým objemem nádoby $V$. Čím větší nádoba, tím více možných míst, kde se molekula může nacházet.
2. **Hybnost / Rychlost (Hybnostní mikrostavy):** Jsou dány teplotou $T$ (potažmo vnitřní energií). Čím vyšší teplota, tím větší je rozptyl rychlostí a tím více energetických hladin (mikrostavů) mají molekuly k dispozici.

Celkový počet mikrostavů $\Omega$ je zhruba řečeno součinem těchto dvou faktorů (dostupného objemu a dostupné hybnosti).

### Analýza Carnotova cyklu pomocí $S = k_B \ln \Omega$

#### 1. Vratná izotermická expanze ($2 \to 3$)

* **Makroskopicky:** Plyn přijímá teplo $Q_H$ od horké lázně při konstantní teplotě $T_H$. Objem roste.
* **Mikroskopicky (Boltzmann):** Protože teplota $T_H$ je přísně konstantní, průměrná kinetická energie a rozptyl rychlostí molekul se nijak nemění. *Hybnostní mikrostavy se tedy nemění.* Plyn ale expanduje (objem $V$ roste), takže molekuly získávají mnohem více prostoru, kde se mohou nacházet. Počet možných prostorových uspořádání masivně roste.
* **Závěr:** Protože se zvětšil prostor pro molekuly, celkový počet mikrostavů vzrostl ($\Omega_2 > \Omega_1$). Podle rovnice $S = k_B \ln \Omega$ tedy **entropie roste** ($\Delta S > 0$).

#### 2. Vratná adiabatická expanze ($3 \to 4$)

* **Makroskopicky:** Plyn je tepelně izolován ($Q = 0$) a dál expanduje. Protože koná práci, jeho teplota klesá z $T_H$ na $T_C$.
* **Mikroskopicky (Boltzmann):** Zde dochází ke rvačce dvou jevů. Plyn dále zvětšuje svůj objem $V$, takže *prostorových mikrostavů přibývá*. Zároveň se ale plyn ochlazuje (klesá $T$), molekuly zpomalují a "smrskává" se paleta možných rychlostí, takže *hybnostních mikrostavů ubývá*.
* Kouzlo dokonale vratného adiabatického děje spočívá v tom, že (podle Liouvillova teorému o zachování fázového objemu) nárůst prostorových možností **přesně vykompenzuje** úbytek těch hybnostních.
* **Závěr:** Celkový počet možných mikrostavů se vůbec nezmění ($\Omega_3 = \Omega_2$). Entropie tak **zůstává konstantní** ($\Delta S = 0$).

#### 3. Vratná izotermická komprese ($4 \to 1$)

* **Makroskopicky:** Plyn odevzdává teplo $Q_C$ studené lázni při stálé teplotě $T_C$. Nádoba se zmenšuje (stlačujeme plyn).
* **Mikroskopicky (Boltzmann):** Teplota $T_C$ je konstantní, takže rozptyl rychlostí zůstává stejný. Objem $V$ se ale zmenšuje. Molekuly jsou natlačovány do stále menšího a menšího prostoru. Odebíráme jim "polohovou svobodu", takže počet možných prostorových uspořádání drasticky klesá.
* **Závěr:** Celkový počet mikrostavů klesá ($\Omega_4 < \Omega_3$). Entropie tedy **klesá** ($\Delta S < 0$).

#### 4. Vratná adiabatická komprese ($1 \to 2$)

* **Makroskopicky:** Plyn je opět izolován a prudce stlačen zpět do výchozího stavu. Teplota roste z $T_C$ na $T_H$.
* **Mikroskopicky (Boltzmann):** Opět se potkávají dva protichůdné jevy. Objem $V$ se zmenšuje (ubývá prostorových mikrostavů), ale plyn se zahřívá, molekuly zrychlují a získávají širší paletu možných rychlostí (přibývá hybnostních mikrostavů). Protože jde opět o dokonalý vnitřně vratný proces, tyto dva vlivy se beze zbytku vyruší.
* **Závěr:** Celkový počet mikrostavů zůstává konstantní ($\Omega_1 = \Omega_4$). Entropie **zůstává konstantní** ($\Delta S = 0$).

Zatímco makroskopická (Clausiova) definice se stará jen o to, kolik tepla proteklo stěnou válce v poměru k teplotě, Boltzmannův přístup nám ukazuje pravý fyzikální příběh: **Změna entropie v Carnotově cyklu není nic jiného než dýchání fázového prostoru.** Plyn dostane více prostoru, jeho stavy se rozprostřou (entropie roste), a pak je zase "nahnán" zpět do ohrady, čímž se jeho možnosti radikálně omezí (entropie klesá).


### Termodynamická teplota (Kelvinova stupnice)

**Termodynamická teplota** je absolutní teplotní stupnice, která je **zcela nezávislá na fyzikálních vlastnostech jakékoliv konkrétní teploměrné látky** (jako je např. tepelná roztažnost rtuti, změna elektrického odporu platiny nebo tlak ideálního plynu).

Její definice nevychází z konstrukce teploměru, ale přímo ze samotných základů **druhého termodynamického zákona**, konkrétně z vlastností vratných (Carnotových) tepelných cyklů.

#### Jak je definována?

Z druhého Carnotova důsledku víme, že účinnost všech vratných cyklů pracujících mezi dvěma stejnými tepelnými lázněmi je identická a závisí *pouze a jedině* na teplotách těchto lázní. Z toho matematicky vyplývá, že poměr tepla přijatého z horké lázně ($Q_H$) a tepla odevzdaného do chladné lázně ($Q_C$) musí odpovídat poměru jejich teplot.

Termodynamická teplota (označovaná $T$) je tedy definována tímto jednoduchým poměrem:

$$ \frac{T_C}{T_H} = \left( \frac{Q_C}{Q_H} \right)_{\text{rev cycle}} $$

Dvě teploty na Kelvinově stupnici jsou ve stejném poměru jako hodnoty tepla přijatého a odevzdaného vratným cyklem, který mezi těmito teplotami pracuje. V tomto případě teplo ($Q$) v podstatě přebírá roli "teploměrné vlastnosti".

#### Dva základní pilíře stupnice:

1. **Absolutní nula (0 K):**
Z definice výše vyplývá, že teplo odevzdané cyklem ($Q_C$) nemůže být záporné, takže ani teplota $T_C$ nemůže být záporná. Jak se odevzdané teplo $Q_C$ blíží k nule, blíží se k nule i teplota $T_C$. Z toho se vyvozuje, že teplota nula na Kelvinově stupnici (0 K) je **absolutně nejnižší myslitelná teplota**. Proto se jí říká absolutní teplotní stupnice.
2. **Referenční bod (273,16 K):**
Abychom mohli teplotu vyčíslit (a nejen znát její poměr), musíme jeden konkrétní stav v přírodě pevně zafixovat. Mezinárodní dohodou byl jako tento referenční bod zvolen **trojný bod vody** (stav, kdy v dokonalé rovnováze existuje led, kapalná voda a vodní pára). Tomuto bodu byla přiřazena exaktní hodnota **273,16 K**.

Jakákoliv jiná termodynamická teplota $T$ je pak definována vztahem vůči tomuto bodu:


$$ T = 273,16 \left( \frac{Q}{Q_{tp}} \right)_{\text{rev cycle}} $$


*(Kde $Q_{tp}$ je teplo vyměněné s lázní na teplotě trojného bodu vody a $Q$ je teplo vyměněné s lázní na měřené teplotě $T$).*

#### Inženýrská praxe a měření

Protože dokonale vratné (Carnotovy) cykly existují jen v teorii, nelze termodynamickou teplotu přímo "změřit" pomocí Carnotova motoru. V praxi se využívají velmi přesné teploměry (např. plynové teploměry s konstantním objemem), jejichž chování se při extrémně nízkých tlacích (kdy se reálný plyn chová jako ideální) s termodynamickou teplotou dokonale shoduje. Ke standardizaci těchto měření slouží *Mezinárodní teplotní stupnice (např. ITS-90)*.
