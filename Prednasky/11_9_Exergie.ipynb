## Exergie – Měřítko kvality energie

Doposud jsme se v termodynamice dívali na energii dvěma pohledy:

1. **První zákon (Zákon zachování):** Energie se neztrácí, pouze mění formu. Všechny formy energie jsou z hlediska množství rovnocenné (1 Joule práce = 1 Joule tepla).
2. **Druhý zákon (Směr a limity):** Zjistili jsme, že teplo nelze stoprocentně přeměnit na práci. Práce je tedy jakási "ušlechtilejší" forma energie než teplo.

Inženýrská praxe ale potřebuje nástroj, který tyto dva zákony spojí do jednoho praktického čísla. Potřebujeme vědět, nejen *kolik* energie máme, ale *jak moc je užitečná*. K tomu slouží **exergie**.

### Definice exergie a "Mrtvý stav"

> **Exergie (Dostupná energie / Maximální užitečná práce)** je absolutně největší možné množství užitečné práce, které můžeme z daného systému (nebo toku energie) získat, pokud ho necháme vratně dojít do úplné termodynamické rovnováhy s okolním prostředím.

Abychom mohli exergii definovat, musíme si nejdříve definovat naše okolí, tzv. **mrtvý stav (Dead state)**.
Představte si naši atmosféru. Má obrovskou tepelnou kapacitu, stálou teplotu (např. $T_0 = 293 \text{ K}$ neboli 20 °C) a stálý tlak ($p_0 = 101,3 \text{ kPa}$).

* Pokud má náš systém stejnou teplotu a tlak jako okolí, je v mrtvém stavu. Nemůžeme z něj získat už vůbec žádnou práci (jeho exergie je **nula**).
* Čím více se stav našeho systému liší od mrtvého stavu (má vyšší/nižší tlak, vyšší/nižší teplotu), tím vyšší má exergii.

Základní rozdíl mezi energií a exergií:

* **Energie se zachovává.** (Zákon zachování energie).
* **Exergie se ničí!** Každý reálný, nevratný proces (tření, míchání, přenos tepla přes rozdíl teplot) exergii nenávratně ničí. Tomuto jevu říkáme **ztráta exergie** (nebo také produkce entropie).

### Inženýrský příklad: Proč není teplo jako teplo?

Představte si, že pracujete jako energetik v průmyslovém podniku a máte k dispozici **1000 kJ tepelné energie** ze dvou různých zdrojů:

1. **Zdroj A:** Odpadní teplo z chladicí vody o teplotě **40 °C**.
2. **Zdroj B:** Spaliny z pece o teplotě **1000 °C**.

Okolní vzduch (náš mrtvý stav) má teplotu **20 °C**.

Z hlediska prvního termodynamického zákona (množství) jsou oba zdroje naprosto rovnocenné. Oba vám dají 1000 kJ energie, kterou můžete použít například k ohřátí jiné látky. Z hlediska druhého zákona je ale jejich hodnota (exergie) diametrálně odlišná.

Pojďme si vypočítat maximální možnou práci, kterou bychom z těchto 1000 kJ tepla mohli dostat, pokud bychom na ně připojili dokonalý (Carnotův) tepelný motor pracující proti okolí ($T_0 = 293 \text{ K}$).

Exergie tepla ($Ex_Q$) se vypočítá jako teplo vynásobené Carnotovou účinností vůči okolí:


$$ Ex_Q = Q \cdot \left( 1 - \frac{T_0}{T} \right) $$

**Analýza Zdroje A (voda 40 °C $\to$ 313 K):**


$$ Ex_{Q,A} = 1000 \cdot \left( 1 - \frac{293}{313} \right) = 1000 \cdot (1 - 0,936) = \mathbf{64 \text{ kJ}} $$


*Výsledek:* Z 1000 kJ energie v této vlažné vodě můžete vyrobit maximálně 64 kJ elektřiny/práce. Zbylých 936 kJ tvoří tzv. *anergie* (bezcenná část energie, která musí být odvedena do okolí).

**Analýza Zdroje B (spaliny 1000 °C $\to$ 1273 K):**


$$ Ex_{Q,B} = 1000 \cdot \left( 1 - \frac{293}{1273} \right) = 1000 \cdot (1 - 0,230) = \mathbf{770 \text{ kJ}} $$


*Výsledek:* Z 1000 kJ energie v horkých spalinách vyrobíte teoreticky až 770 kJ elektřiny!

#### Hlavní inženýrský význam

Tento příklad ukazuje samotné jádro moderní energetiky. Tradiční energetické bilance (např. účinnost plynového kotle 95 %) nám lžou o skutečné efektivitě. Plynový kotel sice přemění 95 % chemické energie plynu na teplo ve vodě (první zákon je spokojen), ale pálíme v něm plyn o obrovské teplotě plamene (cca 1900 °C, obrovská exergie), jen abychom ohřáli vodu v radiátorech na 60 °C (nízká exergie). Většinu "pracovního potenciálu" jsme nenávratně zničili pouhým sražením teploty.

**Exergetická analýza** umožňuje inženýrům najít místa, kde se *skutečně* plýtvá potenciálem (kde se ničí exergie), a navrhovat lepší systémy (např. kogeneraci, kde plyn nejprve vyrobí práci v turbíně a teprve jeho chladnější zbytkové teplo se použije na vytápění).
