# Termodynamika

Cílem této kapitoly je položit pevné základy pro naše budoucí inženýrské analýzy. Než začneme řešit složité rovnice a navrhovat energetické systémy, musíme si přesně definovat jazyk, kterým budeme hovořit, a základní koncepty, o které se budeme opírat.


## Co je to termodynamika a její inženýrské aplikace

**Termodynamika** je vědní obor, který se zabývá energií, jejími přeměnami a vztahy mezi teplem, prací a vlastnostmi látek. Z inženýrského hlediska nás zajímá především to, jak efektivně transformovat energii z jedné formy do druhé, aby sloužila užitečnému účelu.

Inženýrská termodynamika má obrovské množství aplikací, které formují náš moderní svět. Mezi ty nejvýznamnější patří:

* **Energetika a motory:** Automobilové motory, spalovací a parní turbíny, uhelné, plynové a jaderné elektrárny.
* **Chlazení a klimatizace:** Chladničky, tepelná čerpadla, klimatizační systémy (HVAC), chlazení elektroniky a supravodivých systémů.
* **Letectví a kosmonautika:** Proudové motory letadel, raketové pohonné systémy, systémy podpory života na Mezinárodní vesmírné stanici (ISS).
* **Alternativní a moderní energetika:** Palivové články, solární termální systémy, geotermální elektrárny a větrná energetika.
* **Biomedicínské inženýrství:** Umělé orgány, analýza respiračního a oběhového systému člověka.


## Definování systémů

Důležitým prvním krokem při jakékoliv inženýrské analýze je přesně popsat, co vlastně studujeme. V mechanice je typickým prvním krokem definování uvolněného tělesa a identifikace všech sil, které na něj působí. V termodynamice pro tento účel používáme pojem **systém**.

> **Systém** je jakákoliv část prostoru nebo množství hmoty, kterou jsme si vybrali pro naši analýzu. Může to být něco tak jednoduchého, jako je plyn uzavřený v pístu, nebo tak složitého, jako je celé tělo.

Všechno, co leží mimo náš vybraný systém, se nazývá **okolí** (surroundings). Systém je od svého okolí oddělen **hranicí** (boundary). Tato hranice může být v klidu (např. pevné stěny nádrže) nebo v pohybu (např. pohybující se píst). Interakce mezi systémem a jeho okolím – jako je přenos tepla nebo konání práce – probíhají právě přes tuto hranici a hrají v termodynamice klíčovou roli.

### Typy termodynamických systémů

Podle toho, jak systém interaguje se svým okolím, rozlišujeme dva základní typy:

1. **Uzavřený systém (Kontrolní hmotnost)**

    Uzavřený systém je definován pevně daným množstvím hmoty.

    * **Klíčová vlastnost:** Hranicí uzavřeného systému nemůže prostupovat hmota (hmotnost systému je konstantní).
    * **Energie:** Teplo a práce mohou hranici prostupovat.
        * *Speciální případ:* Pokud systém neinteraguje s okolím vůbec (nepřijímá ani neodevzdává hmotu ani energii), nazývá se **izolovaný systém**.

2. **Otevřený systém (Kontrolní objem)**
Často je mnohem jednodušší analyzovat zařízení, kterými hmota protéká (např. turbíny, kompresory, čerpadla, motory). V takovém případě volíme jako náš systém určitou oblast v prostoru, kterou nazýváme **kontrolní objem**.

* **Klíčová vlastnost:** Hranicí kontrolního objemu (nazývanou *kontrolní plocha*) může prostupovat hmota i energie.

> **Příklad z praxe:** Představte si vzduchový kompresor připojený k tlakové nádobě. Pokud nás zajímá, jak dlouho musí kompresor běžet, aby tlak v nádrži stoupl na určitou hodnotu, zvolíme jako náš systém (kontrolní objem) celý kompresor i s nádrží – přes hranice nám prochází elektrická práce a dovnitř proudí vzduch z okolí.

## 1.3 Popis systémů a jejich chování

Abychom mohli systémy zkoumat a předvídat, jak se budou chovat, musíme zavést další pojmy.

### Makroskopický a mikroskopický pohled

Termodynamiku můžeme studovat ze dvou různých úhlů pohledu:

* **Makroskopický přístup (Klasická termodynamika):** Zabývá se celkovým chováním systému. Nepotřebujeme znát chování jednotlivých molekul. Tento přístup je přímější, vyžaduje méně matematických komplikací a pro drtivou většinu inženýrských aplikací je naprosto dostačující.
* **Mikroskopický přístup (Statistická termodynamika):** Vychází přímo ze struktury hmoty. Popisuje průměrné chování obrovského množství částic pomocí statistiky. Tento přístup je nezbytný u speciálních aplikací, jako jsou lasery, plazma, proudění vysoce zředěných plynů nebo extrémně nízké teploty (kryogenika).

### Vlastnost, stav a děj

* **Vlastnost (Property):** Je to jakákoliv makroskopická charakteristika systému (např. hmotnost, objem, tlak, teplota, energie), které lze v daném čase přiřadit číselnou hodnotu, aniž bychom museli znát předchozí historii systému. *Poznámka: Teplo a práce nejsou vlastnosti, protože závisí na tom, jak proces probíhal.*
* **Stav (State):** Odkazuje na momentální kondici systému plně popsanou jeho termodynamickými vlastnostmi.
* **Děj / Proces (Process):** Pokud se změní byť jen jediná vlastnost systému, říkáme, že systém prošel změnou stavu neboli termodynamickým dějem.
* **Ustálený stav (Steady state):** Stav, při kterém se žádná z vlastností systému nemění v čase.
* **Termodynamický cyklus:** Sled dějů, který začíná a končí ve stejném stavu (např. oběh vody v parní elektrárně). Po dokončení cyklu nedochází v systému k žádné čisté změně stavu.

### Extenzivní a intenzivní vlastnosti

Vlastnosti systému rozdělujeme do dvou skupin:

1. **Extenzivní vlastnosti:** Jejich hodnota závisí na velikosti (rozsahu) systému. Jsou aditivní. Příkladem je celková hmotnost, celkový objem nebo celková energie.
2. **Intenzivní vlastnosti:** Jejich hodnota nezávisí na velikosti systému. Příkladem je teplota, tlak nebo hustota.

> **Jak to rozeznat?** Představte si systém s určitou teplotou a hmotností a pomyslně ho rozdělte na polovinu. Každá polovina bude mít poloviční hmotnost (extenzivní vlastnost), ale obě budou mít stále stejnou teplotu (intenzivní vlastnost).

### Fáze a čistá látka

* **Fáze:** Fází označujeme množství hmoty, které je zcela homogenní jak chemickým složením, tak fyzikální strukturou (vše je pevné, nebo vše je kapalné, nebo vše je plynné).
* **Čistá látka:** Látka, která má ve všech svých částech neměnné a stejné chemické složení (např. čistá voda). Čistá látka může existovat ve více fázích současně (např. směs ledu a vody je stále čistá látka, protože chemicky jde o $H_2O$).

### Rovnováha (Equilibrium)

Klasická termodynamika klade velký důraz na rovnovážné stavy. Stav **úplné termodynamické rovnováhy** znamená, že v systému nejsou žádné nevyvážené síly, neexistují žádné teplotní gradienty (tepelná rovnováha), nedochází k fázovým změnám (fázová rovnováha) ani k chemickým reakcím (chemická rovnováha). Pokud izolujeme systém v rovnováze od okolí, nebudou se jeho vlastnosti v čase nijak měnit.

### Skutečné a kvazistatické (kvazirovnovážné) děje

Při skutečných procesech probíhajících v reálném čase se systém většinou v rovnováze nenachází (teplota a tlak se mohou lokálně lišit). Pro zjednodušení analýzy však inženýři často používají idealizaci zvanou **kvazistatický (nebo kvazirovnovážný) děj**.

Při tomto ději se předpokládá, že proces probíhá tak pomalu, že se systém v každém okamžiku nachází jen infinitezimálně (nekonečně málo) daleko od stavu rovnováhy. Ačkoliv jde o idealizaci, umožňuje nám tento koncept aplikovat jednoduché matematické modely na velmi složité reálné problémy.
