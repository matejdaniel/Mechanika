# Klikový mechanismus

Klikové mechanismy jsou technická zařízení umožňující změnu otáčivého pohybu na přímočarý vratný nebo obráceně změnu přímočarého vratného pohybu na otáčivý pohyb.

Rozeznáváme dva druhy klikových mechanismů:

* **Zkrácený** – používá se u jednočinných strojů, kde tlak pracovní látky působí pouze na jednu stranu pístu, např. u spalovacích motorů.
* **Úplný** – používá se u dvojčinných strojů, kde tlak pracovní látky působí střídavě na obě strany pístu, např. dvoučinných čerpadel.

Zkrácený klikový mechanismus se skládá z:
1. pístu,
2. pístního čepu,
3. ojnice,
4. kliky
