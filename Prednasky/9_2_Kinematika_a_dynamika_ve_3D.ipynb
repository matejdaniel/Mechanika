# Kinematika a dynamika ve 3D

## Kinematika hmotného bodu

Pohyb bodu je určen změnou jeho souřadnic ve 3D prostoru, t.j. změnou jeho polohového vektoru:

$$
\vec{r} = \vec{r}(t) = [x(t), y(t), z(t)]
$$

Tyto rovnice odpovídají třem stupňům volnosti hmotného bodu.

Pro určení rychlosti a zrychlení můžeme jednoduše určit časovou změnu polohy a rychlosti

$$
\vec{v} = \frac{\mathrm{d} \vec{r}}{\mathrm{d} t}
$$

$$
\vec{a} = \frac{\mathrm{d} \vec{v}}{\mathrm{d} t}
$$


## Kinematika tuhého tělesa

Tuhé těleso v prostoru má pouze šest stupňů volnosti. Jeho poloha je tedy určena šesti souřadnicemi. Například:

- 3 kartézské souřadnice bodu $A$,
- 3 uhly natočení vůči souřadnicové soustavě

- 3 kartézské souřadnice bodu $A$,
- 2 souřadnice určující směr osy $o$ (např. složky jednotkového vektoru $\mathbf{e}_o$ s podmínkou $|\mathbf{e}_o| = 1$),
- 1 úhel natočení $\varphi$.

## Otáčení kolem pevné osy

Pokud bod $A$ a osa $o$ zachovávají v prostoru i v tělese stálou polohu, pak pohyb je určen pouze časovou závislostí úhlu natočení:
$$
\varphi = \varphi(t)
$$

Rychlost libovolného bodu tuhého tělesa se vyjádří pomocí vektorového součinu:

$$
\vec{v} = \vec{\omega} \times \mathbf{r}
$$

kde:

- $\vec{\omega}$ je vektor úhlové rychlosti,
- $\vec{r}$ je vektor od bodu $A$ na ose k danému bodu tělesa.

Velikost rychlosti je:
$$
v = \omega R
$$
kde $R$ je kolmá vzdálenost od osy otáčení.

## Otáčení kolem pevného bodu

Pokud jeden bod tělesa (např. $A$) má po celou dobu pohybu stálou polohu, hovoříme o **otáčení kolem pevného bodu**.

V každém okamžiku lze takové otáčení popsat jako otáčení kolem určité pevné osy. Úhlová rychlost je obecně časově proměnný vektor:
$$
\boldsymbol{\omega} = \boldsymbol{\omega}(t)
\tag{5.14}
$$

## Posuvný pohyb (translace)

Při posuvném pohybu mají všechny body tělesa stejný vektor rychlosti $\mathbf{v}$, a tedy i stejné trajektorie, lišící se pouze počáteční polohou.

## Obecný pohyb – Chaslesova věta

Libovolný pohyb tuhého tělesa lze složit z posuvného pohybu a otáčení kolem pevného bodu. Rychlost libovolného bodu tělesa se určí jako:
$$
\mathbf{v}_X = \mathbf{v}_A + \boldsymbol{\omega} \times \mathbf{r}_{AX}
\tag{5.16}
$$
kde:

- $\mathbf{v}_A$ je rychlost zvoleného bodu $A$,
- $\boldsymbol{\omega}$ je úhlová rychlost otáčení kolem bodu $A$,
- $\mathbf{r}_{AX}$ je vektor z bodu $A$ do bodu $X$.

## Závislost rychlosti na volbě bodu

Zvolíme-li jiný bod $A'$, platí obdobně:
$$
\mathbf{v}_X = \mathbf{v}_{A'} + \boldsymbol{\omega}' \times \mathbf{r}_{A'X}
$$

Vyjádříme rychlost bodu $A'$ pomocí (5.16):
$$
\mathbf{v}_{A'} = \mathbf{v}_A + \boldsymbol{\omega} \times \mathbf{r}_{AA'}
$$

a mezi vektory platí vztah:
$$
\mathbf{r}_{AX} = \mathbf{r}_{AA'} + \mathbf{r}_{A'X}
\tag{5.19}
$$

Dosazením získáme:
$$
\mathbf{v}_X = \mathbf{v}_A + \boldsymbol{\omega} \times (\mathbf{r}_{AA'} + \mathbf{r}_{A'X}) = \mathbf{v}_{A'} + \boldsymbol{\omega} \times \mathbf{r}_{A'X}
$$

Porovnáním rovnic (5.20) a (5.17) dostáváme důležitý výsledek:
$$
\boldsymbol{\omega}' = \boldsymbol{\omega}
$$

> Úhlová rychlost tuhého tělesa nezávisí na volbě bodu, vůči kterému uvažujeme otáčení.

## Kinematická rovnice tuhého tělesa

Rovnice (5.16) je základní rovnicí kinematiky tuhého tělesa. K určení rychlosti libovolného bodu tělesa potřebujeme dvě vektorové funkce:
$$
\mathbf{v}_A(t), \quad \boldsymbol{\omega}(t)
$$

Tyto funkce jsou analogické k rovnicím pohybu hmotného bodu:
- Poloha: $ \mathbf{r}(t) $
- Rychlost: $ \mathbf{v}(t) = \frac{d\mathbf{r}}{dt} $

U tuhého tělesa je rozdíl v tom, že rovnice neurčuje trajektorii bodu, ale pouze jeho okamžitou rychlost.

## Prostorový pohyb tuhého tělesa a rotace ve 3D

Zatímco posuvný (translační) pohyb nebo rotace kolem jedné pevné osy (2D případ) si vystačí s poměrně jednoduchým matematickým popisem, obecný pohyb tuhého tělesa v trojrozměrném prostoru (3D) představuje jednu z nejkomplexnějších, ale zároveň nejdůležitějších kapitol mechaniky. Tento aparát je nezbytný pro návrh robotických ramen, analýzu letu letadel a raket, gyroskopů nebo satelitů na oběžné dráze.

Volné tuhé těleso ve 3D prostoru má **šest stupňů volnosti**: tři pro posuvný pohyb těžiště (translace v osách X, Y, Z) a tři pro rotační pohyb kolem těžiště.


## Tenzor momentu setrvačnosti

Při rotaci kolem pevné osy jsme moment setrvačnosti $I$ brali jako jediné skalární číslo (např. $I = \frac{1}{2}mr^2$ pro válec). V obecném 3D prostoru to ale nestačí. Pokud na těleso působí moment síly $\mathbf{M}$, výsledná úhlová rychlost $\boldsymbol{\omega}$ a moment hybnosti $\mathbf{L}$ **nemusí mít stejný směr**.

Abychom tuto anizotropii (směrovou závislost) popsali, moment setrvačnosti již nemůže být pouhý skalár, ale stává se **tenzorem druhého řádu**, který můžeme zapsat jako symetrickou matici $3 \times 3$:

$$ I = \begin{pmatrix} I_{xx} & -I_{xy} & -I_{xz} \\ -I_{yx} & I_{yy} & -I_{yz} \\ -I_{zx} & -I_{zy} & I_{zz} \end{pmatrix} $$

Kde:

* **$I_{xx}, I_{yy}, I_{zz}$** jsou **osové momenty setrvačnosti**. Představují "odpor" tělesa proti rotaci kolem daných os (integruje se kvadratická vzdálenost hmoty od dané osy).
* **$I_{xy}, I_{xz}, I_{yz}$** jsou **deviační momenty (součiny setrvačnosti)**. Vyjadřují nesymetrii rozložení hmoty vůči souřadnicovým rovinám. Pokud zvolíme osy tak, že odpovídají osám symetrie tělesa, deviační momenty se vynulují. Tyto speciální osy se nazývají **hlavní osy setrvačnosti**.

Vztah mezi momentem hybnosti $\mathbf{L}$ a úhlovou rychlostí $\boldsymbol{\omega}$ má pak maticový tvar:

$$ \mathbf{L} = I \boldsymbol{\omega} $$

Rozepsáno do složek:

$$ \begin{pmatrix} L_x \\ L_y \\ L_z \end{pmatrix} = \begin{pmatrix} I_{xx} & -I_{xy} & -I_{xz} \\ -I_{yx} & I_{yy} & -I_{yz} \\ -I_{zx} & -I_{zy} & I_{zz} \end{pmatrix} \begin{pmatrix} \omega_x \\ \omega_y \\ \omega_z \end{pmatrix} $$



## Eulerovy úhly

Abychom mohli popsat orientaci (natočení) tělesa v prostoru, potřebujeme určit vztah mezi dvěma souřadnicovými systémy:

1. **Globální soustava (pevná):** Referenční soustava spojená s okolím (např. Zemí).
2. **Lokální soustava (tělesová):** Soustava pevně spojená s rotujícím tělesem (ideálně ležící v jeho hlavních osách setrvačnosti).

Leonard Euler dokázal, že jakékoliv natočení tělesa v prostoru lze vyjádřit jako posloupnost **tří po sobě jdoucích rotací kolem definovaných os**. Tyto úhly rotace se nazývají **Eulerovy úhly**.

Nejčastější konvence používaná v klasické mechanice je rotace typu **Z-X-Z**:

1. **Precese ($\phi$ nebo $\alpha$):** Rotace kolem původní osy Z (těleso se otočí v rovině X-Y). Nová osa X se nazývá *uzlová přímka*.
2. **Nutace ($\theta$ nebo $\beta$):** Rotace kolem nové (uzlové) osy X'. Tato rotace "sklopí" osu Z do nové polohy.
3. **Vlastní rotace ($\psi$ nebo $\gamma$):** Rotace kolem nové, již sklopené osy Z''.

*(V letectví se často používá konvence Z-Y-X, známá jako Yaw-Pitch-Roll, tedy zatáčení, klopení a klonění).*

Výsledná rotační matice $R$, která transformuje vektory z tělesové soustavy do globální, se získá maticovým vynásobením tří dílčích rotačních matic:


$$ R = R_z(\alpha) \cdot R_x(\beta) \cdot R_z(\gamma) $$

Tato celková transformační matice $R$ je ortogonální ($R^T = R^{-1}$) a obsahuje goniometrické funkce všech tří úhlů.



## Pohybové rovnice (Newton-Eulerovy rovnice)

Prostorový pohyb tělesa je plně popsán dvěma vektorovými rovnicemi. První řeší translaci těžiště a druhá rotaci kolem něj. Výhodou vektorového a maticového zápisu je jeho kompaktnost.

### 1. Rovnice posuvného pohybu (Newtonova rovnice)

Rychlost změny hybnosti těžiště se rovná výslednici všech vnějších sil působících na těleso.


$$ \sum \mathbf{F} = m \mathbf{a}_c = m \ddot{\mathbf{r}}_c $$


Kde $m$ je hmotnost, $\mathbf{a}_c$ je zrychlení těžiště a $\ddot{\mathbf{r}}_c$ je druhá derivace polohového vektoru těžiště.

### 2. Rovnice rotačního pohybu (Eulerovy rovnice)

Změna momentu hybnosti tělesa v čase se rovná výslednému vnějšímu momentu sil $\mathbf{M}$. Tato rovnice se standardně řeší v **lokální souřadnicové soustavě pevně spojené s tělesem**.

Protože se lokální soustava sama otáčí úhlovou rychlostí $\boldsymbol{\omega}$ vůči inerciálnímu prostoru, derivace momentu hybnosti ($\mathbf{L} = I \boldsymbol{\omega}$) obsahuje dva členy (aplikace Coriolisovy věty o derivaci vektoru v rotující soustavě):

$$ \mathbf{M} = \frac{d\mathbf{L}}{dt} + \boldsymbol{\omega} \times \mathbf{L} $$

Pokud dosadíme tenzor setrvačnosti, získáme slavné **Eulerovy dynamické rovnice** v maticovém a vektorovém tvaru:

$$ \mathbf{M} = I \dot{\boldsymbol{\omega}} + \boldsymbol{\omega} \times (I \boldsymbol{\omega}) $$

Kde:

* $\mathbf{M}$ je vektor výsledného momentu vnějších sil vztažený k těžišti.
* $J$ je tenzor setrvačnosti (který je v tělesové soustavě konstantní maticí).
* $\dot{\boldsymbol{\omega}}$ je úhlové zrychlení.
* Člen $\boldsymbol{\omega} \times (I \boldsymbol{\omega})$ představuje setrvačné (gyroskopické) účinky, které vznikají tím, že těleso rotuje ve 3D prostoru (tyto efekty v 2D rotaci úplně mizí, protože $\boldsymbol{\omega}$ a $\mathbf{L}$ jsou rovnoběžné).

Pokud zvolíme lokální osy tak, že splynou s **hlavními osami setrvačnosti** tělesa, tenzor $I$ se stane diagonální maticí (deviační momenty jsou nula) a Eulerovy rovnice se rozpadnou na tři nelineární diferenciální rovnice:

$$ M_x = I_{xx} \dot{\omega}_x - (I_{yy} - I_{zz}) \omega_y \omega_z $$

$$ M_y = I_{yy} \dot{\omega}_y - (I_{zz} - I_{xx}) \omega_z \omega_x $$

$$ M_z = I_{zz} \dot{\omega}_z - (I_{xx} - I_{yy}) \omega_x \omega_y $$

Tyto rovnice tvoří absolutní základ pro algoritmizaci pohybu těles v moderním simulačním a herním softwaru nebo při programování řídicích jednotek dronů.
